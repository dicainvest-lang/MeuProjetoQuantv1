```python

# =============================================================================
# ██████╗ ██╗   ██╗ █████╗ ███╗   ██╗████████╗    ████████╗██████╗  █████╗ ██████╗ ██╗███╗   ██╗ ██████╗
# ██╔═══██╗██║   ██║██╔══██╗████╗  ██║╚══██╔══╝    ╚══██╔══╝██╔══██╗██╔══██╗██╔══██╗██║████╗  ██║██╔════╝
# ██║   ██║██║   ██║███████║██╔██╗ ██║   ██║          ██║   ██████╔╝███████║██║  ██║██║██╔██╗ ██║██║  ███╗
# ██║▄▄ ██║██║   ██║██╔══██║██║╚██╗██║   ██║          ██║   ██╔══██╗██╔══██║██║  ██║██║██║╚██╗██║██║   ██║
# ╚██████╔╝╚██████╔╝██║  ██║██║ ╚████║   ██║          ██║   ██║  ██║██║  ██║██████╔╝██║██║ ╚████║╚██████╔╝
#  ╚══▀▀═╝  ╚═════╝ ╚═╝  ╚═╝╚═╝  ╚═══╝  ╚═╝          ╚═╝   ╚═╝  ╚═╝╚═╝  ╚═╝╚═════╝ ╚═╝╚═╝  ╚═══╝ ╚═════╝
#
# ==============================================================================
# SISTEMA QUANTITATIVO DE MACHINE LEARNING PARA TRADING
# Versão: 3.0 Professional | Compatível com Google Colab
# ==============================================================================
#
# DESCRIÇÃO:
#   Pipeline completo de Machine Learning para previsão direcional de ativos
#   financeiros usando dados do Yahoo Finance. Inclui feature engineering
#   avançado, múltiplos modelos, ensemble, backtest realista e análise SHAP.
#
# ESTRUTURA:
#   Passo  1 - Instalação e Importação de Bibliotecas
#   Passo  2 - Download de Dados (Yahoo Finance)
#   Passo  3 - Limpeza e Preparação dos Dados
#   Passo  4 - Construção do Alvo (Target)
#   Passo  5 - Feature Engineering Profissional
#   Passo  6 - Seleção de Features
#   Passo  7 - Normalização
#   Passo  8 - PCA (Opcional)
#   Passo  9 - Split Temporal Profissional
#   Passo 10 - Treinamento de Múltiplos Modelos
#   Passo 11 - Hyperparameter Tuning
#   Passo 12 - Ensemble
#   Passo 13 - Avaliação Completa
#   Passo 14 - Backtest Realista
#   Passo 15 - Feature Importance + SHAP
#   Passo 16 - Previsão com Threshold de Probabilidade
#   Passo 17 - Anti-Overfitting
#   Passo 18 - Walk-Forward Optimization
#   Passo 19 - Pipeline Final Automatizado
#   Passo 20 - Previsão para o Próximo Dia (Produção)
#
# ==============================================================================

# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 1 — INSTALAÇÃO E IMPORTAÇÃO DE BIBLIOTECAS           ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

# ─── Instalação automática no Google Colab ────────────────────────────────────
# As bibliotecas abaixo são instaladas automaticamente caso não estejam presentes.
import subprocess, sys

def _instalar_se_ausente(pacotes):
    for pkg in pacotes:
        mod = pkg.split("[")[0].replace("-","_")
        try:
            __import__(mod)
        except ImportError:
            print(f"  [INSTALL] Instalando {pkg}...")
            subprocess.check_call(
                [sys.executable, "-m", "pip", "install", "-q", pkg],
                stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL
            )
            print(f"  [INSTALL] {pkg} instalado.")

_instalar_se_ausente(["yfinance", "ta", "xgboost", "lightgbm", "catboost", "shap", "optuna"])

# Criar diretório de saída (necessário no Google Colab)
import os as _os
_os.makedirs('/mnt/user-data/outputs', exist_ok=True)
# Em Colab puro (sem montagem), salvar localmente e exibir inline
_COLAB_OUTPUT_DIR = '/content/outputs'
_os.makedirs(_COLAB_OUTPUT_DIR, exist_ok=True)

# ─── Imports Padrão ───────────────────────────────────────────────────────────
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from datetime import datetime, timedelta
import os
import time
import json
from copy import deepcopy

# ─── Dados Financeiros ────────────────────────────────────────────────────────
import yfinance as yf

# ─── Indicadores Técnicos ─────────────────────────────────────────────────────
try:
    import ta
    from ta.trend import SMAIndicator, EMAIndicator, MACD, ADXIndicator
    from ta.momentum import RSIIndicator, ROCIndicator, StochasticOscillator
    from ta.volatility import AverageTrueRange, BollingerBands
    from ta.volume import OnBalanceVolumeIndicator, MFIIndicator
    TA_AVAILABLE = True
except ImportError:
    TA_AVAILABLE = False

# ─── Machine Learning ─────────────────────────────────────────────────────────
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import (TimeSeriesSplit, GridSearchCV,
                                     RandomizedSearchCV, cross_val_score)
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, confusion_matrix,
                              classification_report, roc_curve)
from sklearn.feature_selection import SelectKBest, mutual_info_classif
from sklearn.inspection import permutation_importance
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (RandomForestClassifier, VotingClassifier,
                               StackingClassifier)
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Gradient Boosting
try:
    from xgboost import XGBClassifier
    XGB_AVAILABLE = True
except ImportError:
    XGB_AVAILABLE = False
    print("[AVISO] XGBoost não disponível.")

try:
    from lightgbm import LGBMClassifier
    LGB_AVAILABLE = True
except ImportError:
    LGB_AVAILABLE = False
    print("[AVISO] LightGBM não disponível.")

try:
    from catboost import CatBoostClassifier
    CAT_AVAILABLE = True
except ImportError:
    CAT_AVAILABLE = False
    print("[AVISO] CatBoost não disponível.")

# SHAP
try:
    import shap
    SHAP_AVAILABLE = True
except ImportError:
    SHAP_AVAILABLE = False
    print("[AVISO] SHAP não disponível.")

# Optuna
try:
    import optuna
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    OPTUNA_AVAILABLE = True
except ImportError:
    OPTUNA_AVAILABLE = False
    print("[AVISO] Optuna não disponível.")

# Estatística
from scipy import stats
from statsmodels.stats.stattools import durbin_watson
import statsmodels.api as sm

# ─── Configurações Visuais ────────────────────────────────────────────────────
plt.rcParams.update({
    'figure.facecolor': '#0D1117',
    'axes.facecolor': '#161B22',
    'axes.edgecolor': '#30363D',
    'axes.labelcolor': '#C9D1D9',
    'axes.titlecolor': '#F0F6FC',
    'xtick.color': '#8B949E',
    'ytick.color': '#8B949E',
    'text.color': '#C9D1D9',
    'grid.color': '#21262D',
    'grid.linestyle': '--',
    'grid.alpha': 0.5,
    'legend.facecolor': '#161B22',
    'legend.edgecolor': '#30363D',
    'figure.titlesize': 16,
    'axes.titlesize': 13,
    'axes.labelsize': 11,
    'font.family': 'DejaVu Sans',
})

# Paleta de cores institucional
COLORS = {
    'primary':   '#58A6FF',
    'success':   '#3FB950',
    'danger':    '#F85149',
    'warning':   '#D29922',
    'neutral':   '#8B949E',
    'highlight': '#BC8CFF',
    'bg_dark':   '#0D1117',
    'bg_med':    '#161B22',
    'bg_light':  '#21262D',
    'border':    '#30363D',
    'text_main': '#F0F6FC',
    'text_sub':  '#8B949E',
}

print("=" * 70)
print("  SISTEMA QUANTITATIVO DE ML PARA TRADING — Iniciado com Sucesso")
print("=" * 70)
print(f"  Python     : {sys.version.split()[0]}")
print(f"  Pandas     : {pd.__version__}")
print(f"  Numpy      : {np.__version__}")
print(f"  XGBoost    : {'OK' if XGB_AVAILABLE else 'Não instalado'}")
print(f"  LightGBM   : {'OK' if LGB_AVAILABLE else 'Não instalado'}")
print(f"  CatBoost   : {'OK' if CAT_AVAILABLE else 'Não instalado'}")
print(f"  SHAP       : {'OK' if SHAP_AVAILABLE else 'Não instalado'}")
print(f"  Optuna     : {'OK' if OPTUNA_AVAILABLE else 'Não instalado'}")
print("=" * 70)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                    CONFIGURAÇÕES GLOBAIS DO USUÁRIO                     ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# ─── ALTERE APENAS ESTA SEÇÃO PARA PERSONALIZAR O SISTEMA ────────────────────

# ─── Ativo ────────────────────────────────────────────────────────────────────
# Opções populares:
#   Forex   : "EURUSD=X", "GBPUSD=X", "USDJPY=X"
#   Futuros : "GC=F" (Ouro), "CL=F" (Petróleo), "ES=F" (S&P500 Futuro)
#   Índices : "^GSPC" (S&P500), "^IXIC" (NASDAQ), "^BVSP" (Bovespa)
#   Crypto  : "BTC-USD", "ETH-USD"
#   Ações   : "AAPL", "PETR4.SA", "VALE3.SA"
TICKER = "USDJPY=X"

# ─── Período de Dados ─────────────────────────────────────────────────────────
ANOS_HISTORICO = 10          # Quantidade de anos para baixar
TIMEFRAME = "1d"             # Timeframe: "1d", "1wk", "1mo"

# ─── Configurações do Modelo ──────────────────────────────────────────────────
USAR_PCA = False             # True = reduz dimensionalidade com PCA
N_COMPONENTES_PCA = 20       # Número de componentes PCA (se USAR_PCA=True)
PROB_THRESHOLD = 1.0        # Threshold mínimo de probabilidade para operar
TRAIN_RATIO = 0.70           # 70% treino
VAL_RATIO = 0.15             # 15% validação
TEST_RATIO = 0.15            # 15% teste

# ─── Configurações do Backtest ────────────────────────────────────────────────
CAPITAL_INICIAL = 100_000    # Capital inicial em USD
SPREAD_PIPS = 0.0002         # Spread (0.02% = 2 pips aprox.)
SLIPPAGE = 0.0001            # Slippage estimado
CUSTO_OPERACIONAL = 0.0001   # Custo por operação
RISCO_POR_TRADE = 0.02       # Risco 2% do capital por trade
STOP_LOSS_PCT = 0.02         # Stop Loss em %
TAKE_PROFIT_PCT = 0.04       # Take Profit em %

# ─── Walk-Forward ─────────────────────────────────────────────────────────────
WF_N_SPLITS = 5              # Número de janelas no walk-forward
WF_TEST_SIZE = 60            # Dias de teste em cada janela

print(f"\n  [CONFIG] Ativo    : {TICKER}")
print(f"  [CONFIG] Histórico: {ANOS_HISTORICO} anos | Timeframe: {TIMEFRAME}")
print(f"  [CONFIG] PCA      : {'Ativado (' + str(N_COMPONENTES_PCA) + ' comp.)' if USAR_PCA else 'Desativado'}")
print(f"  [CONFIG] Threshold: {PROB_THRESHOLD:.0%} | Risco/Trade: {RISCO_POR_TRADE:.0%}")


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║               PASSO 2 — DOWNLOAD DE DADOS DO YAHOO FINANCE              ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def baixar_dados(ticker: str, anos: int = 10, interval: str = "1d") -> pd.DataFrame:
    """
    Baixa dados OHLCV do Yahoo Finance via yfinance.

    Parâmetros
    ----------
    ticker   : str  – Símbolo do ativo (ex: "^GSPC", "EURUSD=X")
    anos     : int  – Quantidade de anos de histórico
    interval : str  – Timeframe ("1d", "1wk", "1mo")

    Retorno
    -------
    pd.DataFrame com colunas Open, High, Low, Close, Volume
    """
    data_inicio = (datetime.now() - timedelta(days=anos * 365)).strftime("%Y-%m-%d")
    data_fim = datetime.now().strftime("%Y-%m-%d")

    print(f"\n  [DOWNLOAD] Baixando {ticker} — {data_inicio} a {data_fim} ({interval})")

    df = yf.download(
        ticker,
        start=data_inicio,
        end=data_fim,
        interval=interval,
        auto_adjust=True,   # Ajusta splits e dividendos automaticamente
        progress=False
    )

    # Normalizar colunas — yfinance 0.2.x retorna MultiIndex com Ticker nível 1
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    # Garantir colunas padrão (str + capitalize, ignora NaN de MultiIndex vazio)
    df.columns = [str(c).strip().capitalize() for c in df.columns]

    # Remover colunas duplicadas ou vazias após flatten
    df = df.loc[:, ~df.columns.duplicated()]
    df.columns = [c if c != 'Nan' else '_drop' for c in df.columns]
    df.drop(columns=[c for c in df.columns if c == '_drop'], errors='ignore', inplace=True)

    # Renomear variações de 'Adj close'
    for alt in ['Adj close', 'Adj_close', 'Adjclose']:
        if alt in df.columns and 'Close' not in df.columns:
            df.rename(columns={alt: 'Close'}, inplace=True)

    print(f"  [DOWNLOAD] {len(df):,} barras baixadas | "
          f"{df.index[0].date()} → {df.index[-1].date()}")
    print(f"  [DOWNLOAD] Colunas: {list(df.columns)}")

    return df

# ─── Executar download ────────────────────────────────────────────────────────
df_raw = baixar_dados(TICKER, ANOS_HISTORICO, TIMEFRAME)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 3 — LIMPEZA E PREPARAÇÃO DOS DADOS                   ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def limpar_dados(df: pd.DataFrame) -> pd.DataFrame:
    """
    Realiza a limpeza e validação do DataFrame bruto.

    Operações:
    - Remove duplicatas
    - Ordena por índice temporal (crucial para evitar data leakage)
    - Remove linhas com Close = 0 ou NaN
    - Preenche lacunas de volume com 0
    - Reporta estatísticas de qualidade
    """
    print("\n" + "─" * 60)
    print("  PASSO 3 — Limpeza e Preparação")
    print("─" * 60)

    df = df.copy()
    n_original = len(df)

    # 1. Ordenar temporalmente (CRÍTICO para evitar data leakage)
    df = df.sort_index()

    # 2. Remover duplicatas de índice
    df = df[~df.index.duplicated(keep='last')]

    # 3. Remover preços inválidos (zero ou negativos)
    mask_validos = (df['Close'] > 0) & (df['Open'] > 0)
    df = df[mask_validos]

    # 4. Tratar volume ausente (Forex não tem volume real)
    if 'Volume' in df.columns:
        df['Volume'] = df['Volume'].fillna(0).clip(lower=0)
    else:
        df['Volume'] = 0

    # 5. Remover NaN restantes nas colunas OHLC
    df.dropna(subset=['Open', 'High', 'Low', 'Close'], inplace=True)

    # 6. Garantir tipos corretos
    for col in ['Open', 'High', 'Low', 'Close', 'Volume']:
        if col in df.columns:
            df[col] = df[col].astype(float)

    # Relatório
    removidos = n_original - len(df)
    print(f"  Registros originais : {n_original:,}")
    print(f"  Removidos (inválid.): {removidos:,}")
    print(f"  Registros finais    : {len(df):,}")
    print(f"  Período             : {df.index[0].date()} → {df.index[-1].date()}")
    print(f"  NaN restantes       : {df.isnull().sum().sum()}")
    print(f"  Close médio         : {df['Close'].mean():.4f}")
    print(f"  Volume médio        : {df['Volume'].mean():,.0f}")

    return df

df = limpar_dados(df_raw)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║             PASSO 4 — CONSTRUÇÃO DO ALVO (TARGET VARIABLE)              ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Lookahead Bias:
#   O erro mais comum em sistemas de ML para trading é usar informação FUTURA
#   para tomar decisões no PRESENTE. Para evitar isso:
#   - O target sempre usa o retorno do DIA SEGUINTE (shift(-1))
#   - Features são calculadas apenas com dados passados
#   - O shift garante que no momento t, só temos info até t-1

def construir_target(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cria a variável alvo e features de retorno.

    Target binário:
        1 = Close amanhã > Close hoje  (Alta)
        0 = Close amanhã <= Close hoje (Queda)

    IMPORTANTE: shift(-1) move o retorno FUTURO para a linha presente.
    Isso significa que na linha do dia t, temos o alvo referente a t+1.
    """
    print("\n" + "─" * 60)
    print("  PASSO 4 — Construção do Alvo")
    print("─" * 60)

    df = df.copy()

    # Retorno futuro logarítmico (mais estável para séries longas)
    df['retorno_futuro_log'] = np.log(df['Close'].shift(-1) / df['Close'])

    # Retorno futuro percentual (mais intuitivo)
    df['retorno_futuro_pct'] = df['Close'].pct_change().shift(-1) * 100

    # Retorno em pontos/pips absolutos
    df['retorno_futuro_pts'] = df['Close'].shift(-1) - df['Close']

    # Alvo binário principal (1 = Alta, 0 = Queda)
    df['target'] = (df['retorno_futuro_log'] > 0).astype(int)

    # Alvo de direção numérica (-1, 0, 1) — útil para análise
    df['direcao_futura'] = np.sign(df['retorno_futuro_log'])

    # Retorno presente (para referência e features)
    df['retorno_atual'] = np.log(df['Close'] / df['Close'].shift(1))

    # Remover a última linha (sem target válido)
    df = df.iloc[:-1]

    # Estatísticas do target
    dist = df['target'].value_counts(normalize=True)
    print(f"  Distribuição do target:")
    print(f"    Alta  (1): {dist.get(1, 0):.2%}")
    print(f"    Queda (0): {dist.get(0, 0):.2%}")
    print(f"  Retorno médio futuro: {df['retorno_futuro_pct'].mean():.3f}%")
    print(f"  Volatilidade retorno: {df['retorno_futuro_pct'].std():.3f}%")
    print(f"  Registros com target: {len(df):,}")

    return df

df = construir_target(df)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║         PASSO 5 — FEATURE ENGINEERING PROFISSIONAL                      ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Feature Engineering:
#   Em séries temporais financeiras, o sinal bruto (OHLCV) raramente é
#   suficiente. Precisamos transformar esses dados em representações que
#   capturem padrões de momentum, reversão à média, volatilidade e regime
#   de mercado. Mais features não é sempre melhor — qualidade > quantidade.

def engenharia_de_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cria mais de 100 features quantitativas para trading.

    Categorias:
        1. Médias Móveis (SMA, EMA, VWAP)
        2. Momentum (RSI, ROC, Stochastic, MOM)
        3. Volatilidade (ATR, Bollinger, Desvio Padrão)
        4. Tendência (MACD, ADX, DI)
        5. Mean Reversion (Z-score, distância da média)
        6. Volume (OBV, Money Flow, Volume Relativo)
        7. Candlestick (corpo, sombras, gaps, direção)
        8. Lags (retornos defasados, vol defasada)
        9. Estatísticas Rolling (Skew, Kurt, Correlação)
       10. Regimes de Mercado
    """
    print("\n" + "─" * 60)
    print("  PASSO 5 — Feature Engineering")
    print("─" * 60)

    df = df.copy()

    o = df['Open']
    h = df['High']
    l = df['Low']
    c = df['Close']
    v = df['Volume'].replace(0, np.nan)  # NaN para evitar divisão por zero

    # ── 1. MÉDIAS MÓVEIS ────────────────────────────────────────────────────
    for p in [5, 10, 20, 50, 100, 200]:
        df[f'sma_{p}'] = c.rolling(p).mean()
        df[f'ema_{p}'] = c.ewm(span=p, adjust=False).mean()

    # Razão preço/média (posição relativa)
    for p in [5, 10, 20, 50, 200]:
        df[f'preco_sma_{p}_ratio'] = c / df[f'sma_{p}'] - 1

    # Cruzamento de médias (golden/death cross)
    df['cruz_sma_5_20']  = (df['sma_5'] > df['sma_20']).astype(int)
    df['cruz_sma_20_50'] = (df['sma_20'] > df['sma_50']).astype(int)
    df['cruz_sma_50_200']= (df['sma_50'] > df['sma_200']).astype(int)

    # VWAP aproximado (intraday simplificado para daily)
    typical_price = (h + l + c) / 3
    df['vwap'] = (typical_price * v).rolling(20).sum() / v.rolling(20).sum()
    df['preco_vwap_ratio'] = c / df['vwap'] - 1

    # ── 2. MOMENTUM ─────────────────────────────────────────────────────────
    # RSI (Relative Strength Index)
    for p in [7, 14, 21]:
        delta = c.diff()
        gain = delta.clip(lower=0).rolling(p).mean()
        loss = (-delta.clip(upper=0)).rolling(p).mean()
        rs = gain / loss.replace(0, np.nan)
        df[f'rsi_{p}'] = 100 - (100 / (1 + rs))
        df[f'rsi_{p}_norm'] = df[f'rsi_{p}'] / 50 - 1  # Normalizado [-1, 1]

    # ROC (Rate of Change)
    for p in [5, 10, 20, 60]:
        df[f'roc_{p}'] = c.pct_change(p) * 100

    # Momentum puro
    for p in [5, 10, 20]:
        df[f'mom_{p}'] = c - c.shift(p)
        df[f'mom_{p}_norm'] = df[f'mom_{p}'] / c.shift(p)

    # Estocástico
    for p in [14, 21]:
        low_min  = l.rolling(p).min()
        high_max = h.rolling(p).max()
        denom    = high_max - low_min
        df[f'stoch_k_{p}'] = 100 * (c - low_min) / denom.replace(0, np.nan)
        df[f'stoch_d_{p}'] = df[f'stoch_k_{p}'].rolling(3).mean()
        df[f'stoch_diff_{p}'] = df[f'stoch_k_{p}'] - df[f'stoch_d_{p}']

    # ── 3. VOLATILIDADE ──────────────────────────────────────────────────────
    # ATR (Average True Range)
    tr1 = h - l
    tr2 = (h - c.shift(1)).abs()
    tr3 = (l - c.shift(1)).abs()
    true_range = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
    for p in [7, 14, 21]:
        df[f'atr_{p}'] = true_range.ewm(span=p, adjust=False).mean()
        df[f'atr_{p}_norm'] = df[f'atr_{p}'] / c  # ATR normalizado pelo preço

    # Desvio padrão dos retornos
    ret = c.pct_change()
    for p in [5, 10, 20, 60]:
        df[f'vol_{p}'] = ret.rolling(p).std() * np.sqrt(252)  # Anualizada

    # Bollinger Bands
    for p in [20]:
        mid = c.rolling(p).mean()
        std = c.rolling(p).std()
        df[f'bb_upper_{p}'] = mid + 2 * std
        df[f'bb_lower_{p}'] = mid - 2 * std
        df[f'bb_width_{p}'] = (df[f'bb_upper_{p}'] - df[f'bb_lower_{p}']) / mid
        df[f'bb_pos_{p}']   = (c - df[f'bb_lower_{p}']) / (df[f'bb_upper_{p}'] - df[f'bb_lower_{p}'] + 1e-9)

    # Ratio volatilidade curta/longa (regime de vol)
    df['vol_regime'] = df['vol_5'] / df['vol_20']

    # ── 4. TENDÊNCIA ─────────────────────────────────────────────────────────
    # MACD
    ema12 = c.ewm(span=12, adjust=False).mean()
    ema26 = c.ewm(span=26, adjust=False).mean()
    df['macd']        = ema12 - ema26
    df['macd_signal'] = df['macd'].ewm(span=9, adjust=False).mean()
    df['macd_hist']   = df['macd'] - df['macd_signal']
    df['macd_norm']   = df['macd'] / c  # Normalizado pelo preço

    # ADX + DI (força da tendência)
    for p in [14]:
        dm_plus  = h.diff().clip(lower=0)
        dm_minus = (-l.diff()).clip(lower=0)
        tr_s     = true_range.ewm(span=p, adjust=False).mean()
        di_plus  = 100 * dm_plus.ewm(span=p, adjust=False).mean() / tr_s.replace(0, np.nan)
        di_minus = 100 * dm_minus.ewm(span=p, adjust=False).mean() / tr_s.replace(0, np.nan)
        dx       = 100 * (di_plus - di_minus).abs() / (di_plus + di_minus + 1e-9)
        df[f'adx_{p}']       = dx.ewm(span=p, adjust=False).mean()
        df[f'di_plus_{p}']   = di_plus
        df[f'di_minus_{p}']  = di_minus
        df[f'di_diff_{p}']   = di_plus - di_minus

    # Inclinação da EMA (proxy da tendência)
    for p in [20, 50]:
        df[f'ema_{p}_slope'] = df[f'ema_{p}'].diff(5) / df[f'ema_{p}'].shift(5)

    # ── 5. MEAN REVERSION ────────────────────────────────────────────────────
    # Z-score do preço em relação à média
    for p in [20, 60]:
        mu = c.rolling(p).mean()
        sg = c.rolling(p).std()
        df[f'zscore_{p}'] = (c - mu) / sg.replace(0, np.nan)

    # Distância da EMA (%)
    for p in [20, 50, 200]:
        df[f'dist_ema_{p}'] = (c / df[f'ema_{p}'] - 1) * 100

    # ── 6. VOLUME ────────────────────────────────────────────────────────────
    # Volume relativo (múltiplo do volume médio)
    df['vol_rel_20'] = v / v.rolling(20).mean()
    df['vol_rel_50'] = v / v.rolling(50).mean()

    # OBV (On-Balance Volume)
    obv = (np.sign(c.diff()) * v).fillna(0).cumsum()
    df['obv'] = obv
    df['obv_sma20'] = obv.rolling(20).mean()
    df['obv_ratio'] = obv / (df['obv_sma20'] + 1e-9)

    # Money Flow (CMF simplificado)
    mf_vol = ((c - l) - (h - c)) / (h - l + 1e-9) * v
    df['mf_20'] = mf_vol.rolling(20).sum() / v.rolling(20).sum()

    # Preço × Volume (pressão)
    df['price_vol'] = (c * v).rolling(10).mean()
    df['price_vol_norm'] = df['price_vol'] / df['price_vol'].rolling(60).mean()

    # ── 7. CANDLESTICK ────────────────────────────────────────────────────────
    # Tamanho do corpo
    df['corpo'] = (c - o).abs()
    df['corpo_pct'] = df['corpo'] / o.clip(lower=1e-9)

    # Sombras
    df['sombra_sup'] = h - pd.concat([c, o], axis=1).max(axis=1)
    df['sombra_inf'] = pd.concat([c, o], axis=1).min(axis=1) - l
    df['sombra_sup_pct'] = df['sombra_sup'] / o.clip(lower=1e-9)
    df['sombra_inf_pct'] = df['sombra_inf'] / o.clip(lower=1e-9)

    # Range total
    df['range'] = h - l
    df['range_pct'] = df['range'] / o.clip(lower=1e-9)

    # Gap de abertura
    df['gap'] = (o - c.shift(1)) / c.shift(1).clip(lower=1e-9)

    # Direção do candle (1=alta, -1=baixa, 0=neutro)
    df['direcao_candle'] = np.sign(c - o)

    # Últimos N candles na mesma direção (sequência)
    for n in [3, 5]:
        direcoes = (c > o).astype(int)
        df[f'sequencia_alta_{n}'] = direcoes.rolling(n).sum()
        df[f'pct_alta_{n}'] = df[f'sequencia_alta_{n}'] / n

    # ── 8. LAGS ───────────────────────────────────────────────────────────────
    # Lags dos retornos
    for lag in [1, 2, 3, 5, 10, 20]:
        df[f'ret_lag_{lag}'] = ret.shift(lag)

    # Lags da volatilidade
    for lag in [1, 5, 10]:
        df[f'vol20_lag_{lag}'] = df['vol_20'].shift(lag)

    # Lags do RSI
    for lag in [1, 3, 5]:
        df[f'rsi14_lag_{lag}'] = df['rsi_14'].shift(lag)

    # Lags do MACD
    for lag in [1, 3]:
        df[f'macd_lag_{lag}'] = df['macd'].shift(lag)

    # ── 9. ESTATÍSTICAS ROLLING ──────────────────────────────────────────────
    for p in [20, 60]:
        df[f'skew_{p}']  = ret.rolling(p).skew()
        df[f'kurt_{p}']  = ret.rolling(p).kurt()
        df[f'maxdd_{p}'] = c.rolling(p).apply(
            lambda x: (x[-1] / x.max() - 1) if len(x) > 0 else 0, raw=True
        )
        df[f'autocorr_{p}'] = ret.rolling(p).apply(
            lambda x: pd.Series(x).autocorr(lag=1) if len(x) > 1 else 0, raw=True
        )

    # ── 10. REGIMES DE MERCADO ────────────────────────────────────────────────
    # Regime de tendência: 1=tendência, 0=lateral
    df['regime_tendencia'] = (df['adx_14'] > 25).astype(int)

    # Regime de volatilidade: 1=alta vol, 0=baixa vol
    df['regime_alta_vol'] = (df['vol_regime'] > 1.2).astype(int)

    # Regime com base na posição em relação às médias
    above_sma50  = (c > df['sma_50']).astype(int)
    above_sma200 = (c > df['sma_200']).astype(int)
    df['regime_bull'] = ((above_sma50 == 1) & (above_sma200 == 1)).astype(int)
    df['regime_bear'] = ((above_sma50 == 0) & (above_sma200 == 0)).astype(int)

    # Hurst Exponent simplificado (rolling) — indica se é trending (>0.5) ou reversivo
    def hurst_rs(ts):
        """R/S Analysis simplificada para calcular o expoente de Hurst."""
        if len(ts) < 10 or ts.std() == 0:
            return 0.5
        mean = ts.mean()
        deviations = ts - mean
        cumdev = np.cumsum(deviations)
        R = cumdev.max() - cumdev.min()
        S = ts.std()
        if S == 0:
            return 0.5
        return np.log(R / S) / np.log(len(ts))

    df['hurst_60'] = ret.rolling(60).apply(hurst_rs, raw=True)

    # ─── Features de data/tempo ───────────────────────────────────────────────
    df['dia_semana']  = df.index.dayofweek     # 0=Segunda, 4=Sexta
    df['dia_mes']     = df.index.day
    df['mes']         = df.index.month
    df['trimestre']   = df.index.quarter
    df['fim_semana']  = (df.index.dayofweek >= 3).astype(int)  # Qui-Sex
    df['inicio_mes']  = (df.index.day <= 5).astype(int)
    df['fim_mes']     = (df.index.day >= 25).astype(int)

    # ─── Relatório ────────────────────────────────────────────────────────────
    n_features = len([c for c in df.columns if c not in
                      ['Open','High','Low','Close','Volume',
                       'target','retorno_futuro_log','retorno_futuro_pct',
                       'retorno_futuro_pts','direcao_futura','retorno_atual']])
    print(f"  Features criadas: {n_features}")

    return df

df = engenharia_de_features(df)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                  PASSO 6 — SELEÇÃO DE FEATURES                          ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

# Colunas que NÃO são features (metadados e target)
COLUNAS_NAO_FEATURE = [
    'Open', 'High', 'Low', 'Close', 'Volume',
    'target', 'retorno_futuro_log', 'retorno_futuro_pct',
    'retorno_futuro_pts', 'direcao_futura', 'retorno_atual'
]

def selecionar_features(df: pd.DataFrame,
                        target_col: str = 'target',
                        max_features: int = 60,
                        correlacao_max: float = 0.95) -> list:
    """
    Seleciona as melhores features usando múltiplos critérios:
    1. Remove features com muitos NaN
    2. Remove features de variância zero
    3. Remove multicolinearidade (correlação > threshold)
    4. Ranking por Mutual Information
    5. Retorna top-N features

    Parâmetros
    ----------
    df              : DataFrame com todas as features
    target_col      : Nome da coluna alvo
    max_features    : Máximo de features a selecionar
    correlacao_max  : Threshold de correlação para remoção

    Retorno
    -------
    list de nomes das features selecionadas
    """
    print("\n" + "─" * 60)
    print("  PASSO 6 — Seleção de Features")
    print("─" * 60)

    # Identificar todas as features candidatas
    features_candidatas = [c for c in df.columns
                           if c not in COLUNAS_NAO_FEATURE]

    df_feats = df[features_candidatas + [target_col]].copy()

    # 1. Remover features com > 30% NaN
    limiar_nan = 0.30
    nan_pct = df_feats.isnull().mean()
    features_ok = nan_pct[nan_pct < limiar_nan].index.tolist()
    features_ok = [f for f in features_ok if f != target_col]
    print(f"  Features antes da filtragem : {len(features_candidatas)}")
    print(f"  Após remoção NaN > 30%      : {len(features_ok)}")

    # Preencher NaN restantes para análise
    df_clean = df_feats[features_ok + [target_col]].dropna()

    # 2. Remover variância zero
    variancias = df_clean[features_ok].var()
    features_ok = variancias[variancias > 1e-10].index.tolist()
    print(f"  Após remoção variância zero : {len(features_ok)}")

    if len(features_ok) == 0:
        print("  [ERRO] Nenhuma feature válida encontrada!")
        return []

    # 3. Remover multicolinearidade (correlação de Pearson)
    corr_matrix = df_clean[features_ok].corr().abs()
    upper = corr_matrix.where(
        np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    )
    # Features com alta correlação com outra já selecionada
    remover_corr = set()
    for col in upper.columns:
        if any(upper[col] > correlacao_max):
            remover_corr.add(col)

    features_ok = [f for f in features_ok if f not in remover_corr]
    print(f"  Após remoção multicolinear  : {len(features_ok)}")

    # 4. Mutual Information Ranking
    X = df_clean[features_ok].values
    y = df_clean[target_col].values

    mi_scores = mutual_info_classif(X, y, random_state=42)
    mi_ranking = pd.Series(mi_scores, index=features_ok).sort_values(ascending=False)

    # 5. Selecionar top-N
    features_selecionadas = mi_ranking.head(max_features).index.tolist()
    print(f"  Features finais selecionadas: {len(features_selecionadas)}")
    print(f"\n  Top-15 por Mutual Information:")
    for i, (feat, score) in enumerate(mi_ranking.head(15).items(), 1):
        print(f"    {i:2d}. {feat:<35} MI = {score:.4f}")

    return features_selecionadas, mi_ranking

FEATURES_SELECIONADAS, MI_RANKING = selecionar_features(df)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                    PASSO 7 — NORMALIZAÇÃO                               ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Scalers:
#   StandardScaler: média=0, desvio=1. Ideal para dados sem outliers extremos.
#                   Usado em Regressão Logística e algoritmos baseados em distância.
#   RobustScaler:   usa mediana e IQR. Resistente a outliers.
#                   Preferível para dados financeiros (fat tails, black swans).
#
# REGRA CRUCIAL — Evitar Data Leakage:
#   O scaler DEVE ser fitted APENAS nos dados de TREINO.
#   Depois, usa-se transform em validação e teste.
#   Nunca fazer fit no dataset inteiro antes do split!

def preparar_dataset(df: pd.DataFrame,
                     features: list,
                     target_col: str = 'target') -> tuple:
    """
    Prepara o dataset final removendo NaN e separando X e y.
    """
    colunas_necessarias = features + [target_col, 'Close',
                                      'retorno_futuro_pct', 'retorno_atual']
    colunas_existentes = [c for c in colunas_necessarias if c in df.columns]

    df_modelo = df[colunas_existentes].dropna()

    X = df_modelo[features].values
    y = df_modelo[target_col].values
    datas = df_modelo.index
    closes = df_modelo['Close'].values
    retornos = df_modelo['retorno_futuro_pct'].values if 'retorno_futuro_pct' in df_modelo.columns else np.zeros(len(df_modelo))

    print(f"\n  Dataset preparado: {X.shape[0]:,} amostras × {X.shape[1]} features")
    return X, y, datas, closes, retornos

X, y, DATAS, CLOSES, RETORNOS = preparar_dataset(df, FEATURES_SELECIONADAS)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                     PASSO 8 — PCA (OPCIONAL)                            ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — PCA (Principal Component Analysis):
#   Transforma features correlacionadas em componentes ortogonais (independentes)
#   ordenados por variância explicada. Reduz dimensionalidade e ruído.
#   DESVANTAGEM: perde interpretabilidade (componentes não têm significado direto).
#   USE quando: muitas features correlacionadas, overfitting presente.
#   NÃO USE quando: interpretabilidade é importante (ex: para SHAP).

def aplicar_pca_scree(X_train: np.ndarray,
                       n_max: int = None) -> tuple:
    """
    Ajusta PCA e plota o Scree Plot para auxiliar na escolha do número de componentes.
    Retorna o PCA fitted.
    """
    if n_max is None:
        n_max = min(X_train.shape[1], X_train.shape[0] - 1)

    pca_full = PCA(n_components=n_max, random_state=42)
    pca_full.fit(X_train)

    var_exp = pca_full.explained_variance_ratio_
    var_cum = np.cumsum(var_exp)

    n_95 = np.searchsorted(var_cum, 0.95) + 1
    n_90 = np.searchsorted(var_cum, 0.90) + 1

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f'PCA — Scree Plot | {TICKER}', fontsize=14, fontweight='bold',
                 color=COLORS['text_main'])

    componentes = range(1, len(var_exp) + 1)
    ax1.bar(componentes, var_exp * 100, color=COLORS['primary'], alpha=0.7)
    ax1.set_title('Variância por Componente', color=COLORS['text_main'])
    ax1.set_xlabel('Componente', color=COLORS['text_sub'])
    ax1.set_ylabel('Variância Explicada (%)', color=COLORS['text_sub'])
    ax1.set_xlim(0, min(50, len(var_exp)) + 1)

    ax2.plot(componentes, var_cum * 100, color=COLORS['primary'],
             linewidth=2, marker='o', markersize=3)
    ax2.axhline(y=90, color=COLORS['warning'], linestyle='--', alpha=0.8, label='90%')
    ax2.axhline(y=95, color=COLORS['danger'], linestyle='--', alpha=0.8, label='95%')
    ax2.axvline(x=n_90, color=COLORS['warning'], linestyle=':', alpha=0.6)
    ax2.axvline(x=n_95, color=COLORS['danger'], linestyle=':', alpha=0.6)
    ax2.text(n_90 + 0.5, 88, f'n={n_90}', color=COLORS['warning'], fontsize=9)
    ax2.text(n_95 + 0.5, 93, f'n={n_95}', color=COLORS['danger'], fontsize=9)
    ax2.set_title('Variância Acumulada', color=COLORS['text_main'])
    ax2.set_xlabel('Número de Componentes', color=COLORS['text_sub'])
    ax2.set_ylabel('Variância Acumulada (%)', color=COLORS['text_sub'])
    ax2.set_xlim(0, min(60, len(var_exp)) + 1)
    ax2.legend()

    plt.tight_layout()
    plt.savefig(f'{_COLAB_OUTPUT_DIR}/pca_scree.png', dpi=120, bbox_inches='tight',
                facecolor=COLORS['bg_dark'])
    plt.show()

    print(f"  PCA: {n_90} componentes explicam 90% | {n_95} componentes explicam 95%")
    return pca_full


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 9 — SPLIT TEMPORAL PROFISSIONAL                      ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Por que NÃO usar train_test_split aleatório em finanças?
#   Dados financeiros são séries temporais: observações futuras DEPENDEM das
#   passadas. Um split aleatório "contamina" o treino com dados do futuro,
#   criando um modelo que parece ótimo no backtest mas falha em produção.
#   A divisão DEVE ser cronológica e NUNCA embaralhar os dados.

def split_temporal(X: np.ndarray, y: np.ndarray, datas, closes, retornos,
                   train_r: float = 0.70,
                   val_r: float = 0.15,
                   test_r: float = 0.15) -> dict:
    """
    Realiza divisão temporal do dataset em treino, validação e teste.

    Estrutura cronológica:
    |──────── TREINO (70%) ────────|── VAL (15%) ──|── TESTE (15%) ──|
    passado ─────────────────────────────────────────────> futuro
    """
    n = len(X)
    n_train = int(n * train_r)
    n_val   = int(n * val_r)

    splits = {
        'X_train': X[:n_train],
        'y_train': y[:n_train],
        'X_val':   X[n_train:n_train + n_val],
        'y_val':   y[n_train:n_train + n_val],
        'X_test':  X[n_train + n_val:],
        'y_test':  y[n_train + n_val:],
        'datas_train':   datas[:n_train],
        'datas_val':     datas[n_train:n_train + n_val],
        'datas_test':    datas[n_train + n_val:],
        'closes_train':  closes[:n_train],
        'closes_val':    closes[n_train:n_train + n_val],
        'closes_test':   closes[n_train + n_val:],
        'retornos_train': retornos[:n_train],
        'retornos_val':   retornos[n_train:n_train + n_val],
        'retornos_test':  retornos[n_train + n_val:],
        'n_train': n_train,
        'n_val': n_val,
        'n_test': n - n_train - n_val,
    }

    print("\n" + "─" * 60)
    print("  PASSO 9 — Split Temporal")
    print("─" * 60)
    print(f"  Treino    : {splits['n_train']:>5,} amostras | "
          f"{splits['datas_train'][0].date()} → {splits['datas_train'][-1].date()}")
    print(f"  Validação : {splits['n_val']:>5,} amostras | "
          f"{splits['datas_val'][0].date()} → {splits['datas_val'][-1].date()}")
    print(f"  Teste     : {splits['n_test']:>5,} amostras | "
          f"{splits['datas_test'][0].date()} → {splits['datas_test'][-1].date()}")

    return splits

SPLITS = split_temporal(X, y, DATAS, CLOSES, RETORNOS, TRAIN_RATIO, VAL_RATIO, TEST_RATIO)

# ─── Normalização (fit APENAS no treino) ─────────────────────────────────────
print("\n  Normalizando features...")

# RobustScaler: resistente a outliers financeiros (fat tails, crises)
scaler = RobustScaler()
X_train_s = scaler.fit_transform(SPLITS['X_train'])    # fit + transform no treino
X_val_s   = scaler.transform(SPLITS['X_val'])          # apenas transform
X_test_s  = scaler.transform(SPLITS['X_test'])         # apenas transform
X_all_s   = scaler.transform(X)                        # todos os dados (para análise)

# ─── PCA (se ativado) ─────────────────────────────────────────────────────────
if USAR_PCA:
    print(f"\n  Aplicando PCA com {N_COMPONENTES_PCA} componentes...")
    pca_obj = aplicar_pca_scree(X_train_s)
    pca_final = PCA(n_components=N_COMPONENTES_PCA, random_state=42)
    X_train_s = pca_final.fit_transform(X_train_s)
    X_val_s   = pca_final.transform(X_val_s)
    X_test_s  = pca_final.transform(X_test_s)
    X_all_s   = pca_final.transform(X_all_s)
    var_exp_total = pca_final.explained_variance_ratio_.sum()
    print(f"  PCA: {N_COMPONENTES_PCA} componentes = {var_exp_total:.2%} da variância")


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║           PASSO 10 — TREINAMENTO DE MÚLTIPLOS MODELOS                   ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def criar_modelos() -> dict:
    """
    Instancia todos os modelos com hiperparâmetros padrão razoáveis.
    Estes serão otimizados no Passo 11.
    """
    modelos = {}

    # Logistic Regression — baseline linear
    modelos['LogReg'] = LogisticRegression(
        C=0.1, max_iter=1000, random_state=42, class_weight='balanced'
    )

    # Decision Tree
    modelos['DecTree'] = DecisionTreeClassifier(
        max_depth=5, min_samples_leaf=20, random_state=42, class_weight='balanced'
    )

    # Random Forest
    modelos['RandForest'] = RandomForestClassifier(
        n_estimators=200, max_depth=6, min_samples_leaf=10,
        max_features='sqrt', random_state=42, class_weight='balanced', n_jobs=-1
    )

    # XGBoost
    if XGB_AVAILABLE:
        modelos['XGBoost'] = XGBClassifier(
            n_estimators=300, max_depth=4, learning_rate=0.05,
            subsample=0.8, colsample_bytree=0.8, gamma=0.1,
            eval_metric='logloss',
            random_state=42, n_jobs=-1, verbosity=0
        )

    # LightGBM
    if LGB_AVAILABLE:
        modelos['LightGBM'] = LGBMClassifier(
            n_estimators=300, max_depth=4, learning_rate=0.05,
            subsample=0.8, colsample_bytree=0.8, min_child_samples=20,
            class_weight='balanced', random_state=42, n_jobs=-1, verbose=-1
        )

    # CatBoost
    if CAT_AVAILABLE:
        modelos['CatBoost'] = CatBoostClassifier(
            iterations=300, depth=4, learning_rate=0.05,
            eval_metric='AUC', auto_class_weights='Balanced',
            random_seed=42, verbose=0
        )

    return modelos

def treinar_avaliar_modelos(modelos: dict,
                             X_train, y_train,
                             X_val, y_val,
                             X_test, y_test) -> pd.DataFrame:
    """
    Treina e avalia todos os modelos, retornando um DataFrame de comparação.
    """
    print("\n" + "─" * 60)
    print("  PASSO 10 — Treinamento e Comparação de Modelos")
    print("─" * 60)

    resultados = []
    modelos_treinados = {}

    for nome, modelo in modelos.items():
        t0 = time.time()
        print(f"  Treinando {nome}...", end=' ', flush=True)

        # Treinar
        modelo.fit(X_train, y_train)

        # Predições
        y_pred_val  = modelo.predict(X_val)
        y_pred_test = modelo.predict(X_test)

        y_prob_val  = modelo.predict_proba(X_val)[:, 1]
        y_prob_test = modelo.predict_proba(X_test)[:, 1]

        # Métricas
        res = {
            'Modelo': nome,
            'Acc_Val':  accuracy_score(y_val, y_pred_val),
            'Acc_Test': accuracy_score(y_test, y_pred_test),
            'F1_Val':   f1_score(y_val, y_pred_val, zero_division=0),
            'F1_Test':  f1_score(y_test, y_pred_test, zero_division=0),
            'ROC_Val':  roc_auc_score(y_val, y_prob_val),
            'ROC_Test': roc_auc_score(y_test, y_prob_test),
            'Prec_Test': precision_score(y_test, y_pred_test, zero_division=0),
            'Rec_Test':  recall_score(y_test, y_pred_test, zero_division=0),
            'Tempo_s':   round(time.time() - t0, 2),
        }
        resultados.append(res)
        modelos_treinados[nome] = {
            'modelo': modelo,
            'y_pred_test': y_pred_test,
            'y_prob_test': y_prob_test,
        }

        print(f"OK | ROC-AUC Test: {res['ROC_Test']:.4f} | "
              f"F1: {res['F1_Test']:.4f} | {res['Tempo_s']}s")

    df_res = pd.DataFrame(resultados).sort_values('ROC_Test', ascending=False)

    print(f"\n  {'─'*60}")
    print(f"  {'Modelo':<15} {'ROC-AUC':<10} {'Acc':<10} {'F1':<10} {'Prec':<10} {'Rec':<10}")
    print(f"  {'─'*60}")
    for _, row in df_res.iterrows():
        print(f"  {row['Modelo']:<15} {row['ROC_Test']:<10.4f} "
              f"{row['Acc_Test']:<10.4f} {row['F1_Test']:<10.4f} "
              f"{row['Prec_Test']:<10.4f} {row['Rec_Test']:<10.4f}")

    return df_res, modelos_treinados

MODELOS = criar_modelos()
DF_RESULTADOS, MODELOS_TREINADOS = treinar_avaliar_modelos(
    MODELOS,
    X_train_s, SPLITS['y_train'],
    X_val_s, SPLITS['y_val'],
    X_test_s, SPLITS['y_test']
)

# ── Selecionar melhor modelo ─────────────────────────────────────────
# Preferência: modelos tree-based (suportam SHAP + feature_importances_)
# Se o top-1 for linear (LogReg), verificar se existe tree-based próximo
# (dentro de 0.5% de ROC-AUC) para ter interpretabilidade completa.
MELHOR_MODELO_NOME = DF_RESULTADOS.iloc[0]['Modelo']
MODELOS_LINEARES = {'LogReg'}
MODELOS_TREE     = {'XGBoost', 'LightGBM', 'CatBoost', 'RandForest',
                    'DecTree', 'XGBoost_Tuned', 'RF_Tuned'}

if MELHOR_MODELO_NOME in MODELOS_LINEARES:
    roc_top = DF_RESULTADOS.iloc[0]['ROC_Test']
    # Buscar o melhor tree-based dentro de 0.5% do top-1
    df_tree = DF_RESULTADOS[DF_RESULTADOS['Modelo'].isin(MODELOS_TREE)]
    if len(df_tree) > 0:
        roc_tree_best = df_tree.iloc[0]['ROC_Test']
        if roc_top - roc_tree_best <= 0.005:
            MELHOR_MODELO_NOME = df_tree.iloc[0]['Modelo']
            print(f"  [AUTO] Trocando para {MELHOR_MODELO_NOME} (tree-based, "
                  f"ROC-AUC {roc_tree_best:.4f} vs {roc_top:.4f}) "
                  f"para habilitar SHAP e Feature Importance.")
        else:
            print(f"  [INFO] Mantendo {MELHOR_MODELO_NOME} (linear): "
                  f"melhor tree-based está {roc_top - roc_tree_best:.4f} abaixo.")

MELHOR_MODELO = MODELOS_TREINADOS[MELHOR_MODELO_NOME]['modelo']
print(f"\n  Melhor modelo final : {MELHOR_MODELO_NOME} "
      f"(ROC-AUC = {DF_RESULTADOS[DF_RESULTADOS['Modelo']==MELHOR_MODELO_NOME]['ROC_Test'].values[0]:.4f})")


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 11 — HYPERPARAMETER TUNING                           ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — TimeSeriesSplit para validação cruzada temporal:
#   Em vez de k-fold aleatório, o TimeSeriesSplit expande a janela de treino
#   progressivamente, garantindo que validação sempre use dados mais recentes.

def tuning_optuna_xgboost(X_train, y_train,
                            n_trials: int = 50) -> dict:
    """
    Otimiza XGBoost com Optuna usando validação temporal.
    Optuna usa Bayesian optimization (TPE) para busca mais eficiente
    que GridSearch/RandomSearch.
    """
    if not (OPTUNA_AVAILABLE and XGB_AVAILABLE):
        print("  [SKIP] Optuna ou XGBoost não disponível.")
        return {}

    print("\n  Otimizando XGBoost com Optuna (TimeSeriesSplit)...")

    tscv = TimeSeriesSplit(n_splits=5)

    def objective(trial):
        params = {
            'n_estimators':     trial.suggest_int('n_estimators', 100, 500),
            'max_depth':        trial.suggest_int('max_depth', 3, 7),
            'learning_rate':    trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
            'subsample':        trial.suggest_float('subsample', 0.6, 1.0),
            'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
            'gamma':            trial.suggest_float('gamma', 0, 0.5),
            'min_child_weight': trial.suggest_int('min_child_weight', 1, 20),
            'reg_alpha':        trial.suggest_float('reg_alpha', 1e-5, 1.0, log=True),
            'reg_lambda':       trial.suggest_float('reg_lambda', 1e-5, 1.0, log=True),
            'eval_metric': 'logloss',
            'random_state': 42,
            'verbosity': 0,
            'n_jobs': -1,
        }
        modelo = XGBClassifier(**params)
        scores = []
        for train_idx, val_idx in tscv.split(X_train):
            modelo.fit(X_train[train_idx], y_train[train_idx])
            proba = modelo.predict_proba(X_train[val_idx])[:, 1]
            scores.append(roc_auc_score(y_train[val_idx], proba))
        return np.mean(scores)

    study = optuna.create_study(direction='maximize',
                                 sampler=optuna.samplers.TPESampler(seed=42))
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    best_params = study.best_params
    best_score  = study.best_value
    print(f"  Melhor ROC-AUC (CV): {best_score:.4f}")
    print(f"  Melhores params: {best_params}")

    return best_params

def tuning_random_search(modelo_nome: str,
                          X_train, y_train,
                          n_iter: int = 30) -> dict:
    """
    RandomizedSearchCV com TimeSeriesSplit para modelos sem Optuna.
    """
    tscv = TimeSeriesSplit(n_splits=5)

    grade_params = {
        'RandForest': {
            'n_estimators': [100, 200, 300],
            'max_depth': [4, 5, 6, 7, None],
            'min_samples_leaf': [5, 10, 20],
            'max_features': ['sqrt', 'log2'],
        },
        'LightGBM': {
            'n_estimators': [100, 200, 300, 500],
            'max_depth': [3, 4, 5, 6],
            'learning_rate': [0.01, 0.05, 0.1],
            'num_leaves': [20, 31, 50],
            'min_child_samples': [10, 20, 30],
        },
    }

    if modelo_nome not in grade_params:
        return {}

    print(f"\n  RandomSearchCV para {modelo_nome}...")
    modelo = MODELOS[modelo_nome]
    rs = RandomizedSearchCV(
        modelo, grade_params[modelo_nome],
        n_iter=n_iter, cv=tscv, scoring='roc_auc',
        n_jobs=-1, random_state=42, verbose=0
    )
    rs.fit(X_train, y_train)
    print(f"  Melhor score CV: {rs.best_score_:.4f}")
    return rs.best_params_

# ─── Executar tuning ──────────────────────────────────────────────────────────
print("\n" + "─" * 60)
print("  PASSO 11 — Hyperparameter Tuning")
print("─" * 60)

BEST_PARAMS_XGB = {}
BEST_PARAMS_RF  = {}

# Optuna para XGBoost (mais eficiente)
if OPTUNA_AVAILABLE and XGB_AVAILABLE:
    BEST_PARAMS_XGB = tuning_optuna_xgboost(X_train_s, SPLITS['y_train'], n_trials=40)

# RandomSearch para Random Forest
if 'RandForest' in MODELOS:
    BEST_PARAMS_RF = tuning_random_search('RandForest', X_train_s, SPLITS['y_train'])

# Retreinar com melhores parâmetros
if BEST_PARAMS_XGB and XGB_AVAILABLE:
    BEST_PARAMS_XGB.update({'eval_metric': 'logloss',
                             'random_state': 42, 'n_jobs': -1, 'verbosity': 0})
    MODELOS['XGBoost_Tuned'] = XGBClassifier(**BEST_PARAMS_XGB)
    MODELOS['XGBoost_Tuned'].fit(X_train_s, SPLITS['y_train'])

if BEST_PARAMS_RF and 'RandForest' in MODELOS:
    MODELOS['RF_Tuned'] = RandomForestClassifier(**BEST_PARAMS_RF,
                                                   random_state=42, n_jobs=-1)
    MODELOS['RF_Tuned'].fit(X_train_s, SPLITS['y_train'])


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                     PASSO 12 — ENSEMBLE                                 ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Ensemble:
#   Combinar múltiplos modelos geralmente supera qualquer modelo individual.
#   Tipos:
#   - Voting: voto por maioria (hard) ou média de probabilidades (soft)
#   - Stacking: usa as predições de modelos base como features de um meta-modelo
#   - Blending ponderado: pesos baseados na performance na validação

def criar_ensemble(modelos_treinados: dict,
                    X_val, y_val,
                    X_train, y_train) -> dict:
    """
    Cria diferentes tipos de ensemble e retorna o melhor.

    CORREÇÃO: usa o parâmetro `modelos_treinados` (local) em toda a função,
    evitando NameError quando executado em células separadas no Colab.
    """
    print("\n" + "─" * 60)
    print("  PASSO 12 — Ensemble")
    print("─" * 60)

    ensembles = {}

    # ── Selecionar os melhores modelos disponíveis ────────────────────────────
    # Prioridade: boosting > árvores > linear (melhor diversidade no ensemble)
    estimadores = []
    for nome in ['XGBoost', 'LightGBM', 'CatBoost', 'RandForest', 'DecTree', 'LogReg']:
        if nome in modelos_treinados:                          # ← parâmetro local
            estimadores.append((nome, modelos_treinados[nome]['modelo']))

    # Fallback: pegar os N primeiros disponíveis
    if len(estimadores) < 2:
        estimadores = [
            (nome, info['modelo'])
            for nome, info in list(modelos_treinados.items())
        ]

    if len(estimadores) < 2:
        print("  [SKIP] Modelos insuficientes para ensemble (mínimo 2).")
        return {}

    print(f"  Modelos no ensemble: {[e[0] for e in estimadores]}")

    # ── Voting Soft (média das probabilidades) ────────────────────────────────
    print("  Criando Voting Classifier (soft)...")
    try:
        voting = VotingClassifier(estimators=estimadores, voting='soft', n_jobs=-1)
        voting.fit(X_train, y_train)
        prob_val = voting.predict_proba(X_val)[:, 1]
        auc_val  = roc_auc_score(y_val, prob_val)
        ensembles['Voting_Soft'] = {'modelo': voting, 'auc_val': auc_val}
        print(f"  Voting Soft — ROC-AUC Val: {auc_val:.4f}")
    except Exception as e:
        print(f"  [AVISO] Voting falhou: {e}")

    # ── Stacking ──────────────────────────────────────────────────────────────
    # Usa no mínimo 2 e no máximo 3 modelos base para evitar IndexError
    base_stacking = estimadores[:min(3, len(estimadores))]
    if len(base_stacking) >= 2:
        print("  Criando Stacking Classifier...")
        try:
            meta_learner = LogisticRegression(C=0.1, max_iter=500, random_state=42)
            # Número de splits ajustado ao tamanho do treino
            # (mínimo 100 amostras por fold para estabilidade)
            n_cv = min(3, max(2, len(X_train) // 100))
            stacking = StackingClassifier(
                estimators=base_stacking,
                final_estimator=meta_learner,
                cv=n_cv,          # int → StratifiedKFold (evita erro de partição)
                n_jobs=-1,
                passthrough=False
            )
            stacking.fit(X_train, y_train)
            prob_val_st = stacking.predict_proba(X_val)[:, 1]
            auc_val_st  = roc_auc_score(y_val, prob_val_st)
            ensembles['Stacking'] = {'modelo': stacking, 'auc_val': auc_val_st}
            print(f"  Stacking       — ROC-AUC Val: {auc_val_st:.4f}")
        except Exception as e:
            print(f"  [AVISO] Stacking falhou: {e}")
    else:
        print("  [SKIP] Stacking requer pelo menos 2 modelos base.")

    # ── Ensemble Ponderado (pesos = ROC-AUC no val) ───────────────────────────
    print("  Criando Ensemble Ponderado...")
    try:
        pesos = {}
        probs_val_pond = np.zeros(len(y_val))
        soma_pesos = 0.0

        for nome, info in modelos_treinados.items():           # ← parâmetro local
            prob = info['modelo'].predict_proba(X_val)[:, 1]
            auc  = roc_auc_score(y_val, prob)
            pesos[nome] = auc
            probs_val_pond += auc * prob
            soma_pesos += auc

        if soma_pesos > 0:
            probs_val_pond /= soma_pesos
            auc_pond = roc_auc_score(y_val, probs_val_pond)
            ensembles['Ponderado'] = {
                'pesos': pesos,
                'soma_pesos': soma_pesos,
                'auc_val': auc_pond
            }
            print(f"  Ponderado      — ROC-AUC Val: {auc_pond:.4f}")
    except Exception as e:
        print(f"  [AVISO] Ensemble ponderado falhou: {e}")

    if not ensembles:
        print("  [SKIP] Nenhum ensemble foi criado com sucesso.")
        return {}

    # Melhor ensemble
    melhor = max(ensembles.items(), key=lambda x: x[1]['auc_val'])
    print(f"\n  Melhor ensemble: {melhor[0]} (ROC-AUC = {melhor[1]['auc_val']:.4f})")

    return ensembles

ENSEMBLES = criar_ensemble(MODELOS_TREINADOS, X_val_s, SPLITS['y_val'],
                            X_train_s, SPLITS['y_train'])


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 13 — AVALIAÇÃO COMPLETA                              ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def metricas_financeiras(retornos_modelo: np.ndarray,
                          retornos_bh: np.ndarray,
                          capital_inicial: float = 100_000) -> dict:
    """
    Calcula métricas financeiras do backtest.

    Retornos esperados em formato decimal (ex: 0.01 = 1%)
    """
    # Equity curve
    equity = capital_inicial * (1 + retornos_modelo).cumprod()
    equity_bh = capital_inicial * (1 + retornos_bh).cumprod()

    # Retorno total
    ret_total = (equity.iloc[-1] / capital_inicial - 1) if len(equity) > 0 else 0
    ret_bh    = (equity_bh.iloc[-1] / capital_inicial - 1) if len(equity_bh) > 0 else 0

    # Sharpe Ratio (anualizado, assumindo 252 dias)
    std = retornos_modelo.std()
    sharpe = (retornos_modelo.mean() / std * np.sqrt(252)) if std > 0 else 0

    # Sortino Ratio (penaliza apenas retornos negativos)
    downside = retornos_modelo[retornos_modelo < 0].std()
    sortino = (retornos_modelo.mean() / downside * np.sqrt(252)) if downside > 0 else 0

    # Drawdown
    roll_max = equity.cummax()
    drawdown = (equity - roll_max) / roll_max
    max_dd = drawdown.min()

    # Calmar Ratio
    ann_ret = ret_total / (len(retornos_modelo) / 252) if len(retornos_modelo) > 0 else 0
    calmar  = (-ann_ret / max_dd) if max_dd < 0 else 0

    # Win Rate — garantir Series para operações booleanas
    rm = pd.Series(retornos_modelo) if not isinstance(retornos_modelo, pd.Series) else retornos_modelo
    rm = rm.dropna()
    wins   = int((rm > 0).sum())
    losses = int((rm <= 0).sum())
    win_rate = wins / len(rm) if len(rm) > 0 else 0

    # Profit Factor
    gross_profit = rm[rm > 0].sum()
    gross_loss   = abs(rm[rm < 0].sum())
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else np.inf

    # Payoff (ganho médio / perda média)
    avg_win  = rm[rm > 0].mean() if wins > 0 else 0
    avg_loss = abs(rm[rm < 0].mean()) if losses > 0 else 1
    payoff   = avg_win / avg_loss if avg_loss > 0 else 0

    # Expectancy
    expectancy = win_rate * avg_win - (1 - win_rate) * avg_loss

    return {
        'Retorno Total':    ret_total,
        'Retorno B&H':      ret_bh,
        'Sharpe Ratio':     sharpe,
        'Sortino Ratio':    sortino,
        'Max Drawdown':     max_dd,
        'Calmar Ratio':     calmar,
        'Win Rate':         win_rate,
        'Profit Factor':    profit_factor,
        'Payoff':           payoff,
        'Expectancy':       expectancy,
        'N Trades':         len(retornos_modelo),
        'equity':           equity,
        'equity_bh':        equity_bh,
        'drawdown':         drawdown,
    }

def avaliar_completo(modelo, nome_modelo: str,
                      X_test, y_test, retornos_test, closes_test,
                      datas_test) -> dict:
    """
    Avaliação completa: classificação + financeira + gráficos.
    """
    print("\n" + "─" * 60)
    print(f"  PASSO 13 — Avaliação Completa | {nome_modelo}")
    print("─" * 60)

    y_pred = modelo.predict(X_test)
    y_prob = modelo.predict_proba(X_test)[:, 1]

    # ── Métricas de Classificação ─────────────────────────────────────────────
    acc   = accuracy_score(y_test, y_pred)
    prec  = precision_score(y_test, y_pred, zero_division=0)
    rec   = recall_score(y_test, y_pred, zero_division=0)
    f1    = f1_score(y_test, y_pred, zero_division=0)
    roc   = roc_auc_score(y_test, y_prob)

    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(f"  F1-Score  : {f1:.4f}")
    print(f"  ROC-AUC   : {roc:.4f}")

    # ── Retornos do modelo ────────────────────────────────────────────────────
    # Estratégia long-short: compra se pred=1, vende se pred=0
    sinais = np.where(y_pred == 1, 1, -1)
    retornos_modelo = pd.Series(sinais * retornos_test / 100,
                                index=datas_test)
    retornos_bh     = pd.Series(retornos_test / 100, index=datas_test)

    # Descontar custos operacionais
    n_trocas = (np.diff(y_pred) != 0).sum()
    custo_total = n_trocas * (SPREAD_PIPS + SLIPPAGE + CUSTO_OPERACIONAL)
    retornos_modelo.iloc[-1] -= custo_total  # simplificação

    metricas_fin = metricas_financeiras(retornos_modelo, retornos_bh, CAPITAL_INICIAL)

    print(f"\n  ── Métricas Financeiras ──")
    print(f"  Retorno Modelo : {metricas_fin['Retorno Total']:.2%}")
    print(f"  Retorno B&H    : {metricas_fin['Retorno B&H']:.2%}")
    print(f"  Sharpe         : {metricas_fin['Sharpe Ratio']:.3f}")
    print(f"  Sortino        : {metricas_fin['Sortino Ratio']:.3f}")
    print(f"  Max Drawdown   : {metricas_fin['Max Drawdown']:.2%}")
    print(f"  Win Rate       : {metricas_fin['Win Rate']:.2%}")
    print(f"  Profit Factor  : {metricas_fin['Profit Factor']:.3f}")
    print(f"  Payoff         : {metricas_fin['Payoff']:.3f}")
    print(f"  Expectancy     : {metricas_fin['Expectancy']:.4f}")
    print(f"  Calmar Ratio   : {metricas_fin['Calmar Ratio']:.3f}")

    return {
        'y_pred': y_pred, 'y_prob': y_prob,
        'acc': acc, 'prec': prec, 'rec': rec, 'f1': f1, 'roc': roc,
        'metricas_fin': metricas_fin,
        'retornos_modelo': retornos_modelo,
        'retornos_bh': retornos_bh,
    }

AVALIACAO = avaliar_completo(
    MELHOR_MODELO, MELHOR_MODELO_NOME,
    X_test_s, SPLITS['y_test'],
    SPLITS['retornos_test'],
    SPLITS['closes_test'],
    SPLITS['datas_test']
)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                   PASSO 14 — BACKTEST REALISTA                          ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def backtest_realista(y_pred: np.ndarray,
                       y_prob: np.ndarray,
                       closes: np.ndarray,
                       retornos: np.ndarray,
                       datas,
                       capital_inicial: float = 100_000,
                       prob_threshold: float = 0.55,
                       risco_trade: float = 0.02,
                       stop_loss: float = 0.02,
                       take_profit: float = 0.04,
                       spread: float = 0.0002,
                       slippage: float = 0.0001) -> dict:
    """
    Backtest realista com:
    - Threshold de probabilidade (só opera sinais com alta confiança)
    - Position sizing por risco
    - Stop-Loss e Take-Profit
    - Spread + Slippage
    """
    print("\n" + "─" * 60)
    print("  PASSO 14 — Backtest Realista")
    print("─" * 60)

    capital = capital_inicial
    equity_curve = [capital]
    trades = []
    posicao = 0  # 0=neutro, 1=long, -1=short

    for i in range(len(y_pred)):
        preco_entrada = closes[i]
        retorno_real  = retornos[i] / 100

        # Filtro de threshold
        sinal = 0
        if y_prob[i] > prob_threshold:
            sinal = 1   # Long
        elif y_prob[i] < (1 - prob_threshold):
            sinal = -1  # Short

        if sinal == 0:
            equity_curve.append(capital)
            continue

        # Custo de entrada (spread + slippage)
        custo_entrada = (spread + slippage) * capital

        # Position sizing por risco
        tamanho_pos = capital * risco_trade

        # P&L bruto
        pnl_bruto = tamanho_pos * sinal * retorno_real

        # Aplicar Stop-Loss / Take-Profit
        if sinal == 1:
            if retorno_real < -stop_loss:
                pnl_bruto = -tamanho_pos * stop_loss
            elif retorno_real > take_profit:
                pnl_bruto = tamanho_pos * take_profit
        else:  # short
            if retorno_real > stop_loss:
                pnl_bruto = -tamanho_pos * stop_loss
            elif retorno_real < -take_profit:
                pnl_bruto = tamanho_pos * take_profit

        # P&L líquido
        pnl_liquido = pnl_bruto - custo_entrada
        capital += pnl_liquido

        equity_curve.append(capital)
        trades.append({
            'data': datas[i],
            'sinal': sinal,
            'prob': y_prob[i],
            'preco': preco_entrada,
            'retorno_real': retorno_real,
            'pnl': pnl_liquido,
            'capital': capital,
        })

    # Métricas
    equity_s = pd.Series(equity_curve, index=list(datas[:len(equity_curve)]) if len(equity_curve) <= len(datas) else list(datas) + [datas[-1]])
    ret_total = (capital / capital_inicial - 1)

    df_trades = pd.DataFrame(trades)
    n_trades = len(df_trades)

    if n_trades > 0:
        wins = (df_trades['pnl'] > 0).sum()
        win_rate = wins / n_trades
        pf_num = df_trades[df_trades['pnl'] > 0]['pnl'].sum()
        pf_den = abs(df_trades[df_trades['pnl'] < 0]['pnl'].sum())
        profit_factor = pf_num / pf_den if pf_den > 0 else np.inf
    else:
        win_rate = profit_factor = 0

    # Drawdown
    eq_series = pd.Series(equity_curve)
    roll_max  = eq_series.cummax()
    dd        = (eq_series - roll_max) / roll_max
    max_dd    = dd.min()

    print(f"  Capital inicial  : ${capital_inicial:>12,.2f}")
    print(f"  Capital final    : ${capital:>12,.2f}")
    print(f"  Retorno total    : {ret_total:.2%}")
    print(f"  N° de trades     : {n_trades:,}")
    print(f"  Win Rate         : {win_rate:.2%}")
    print(f"  Profit Factor    : {profit_factor:.3f}")
    print(f"  Max Drawdown     : {max_dd:.2%}")
    print(f"  Threshold usado  : {prob_threshold:.0%}")

    return {
        'equity_curve': pd.Series(equity_curve),
        'df_trades': df_trades,
        'ret_total': ret_total,
        'n_trades': n_trades,
        'win_rate': win_rate,
        'profit_factor': profit_factor,
        'max_dd': max_dd,
        'capital_final': capital,
    }

BACKTEST = backtest_realista(
    AVALIACAO['y_pred'], AVALIACAO['y_prob'],
    SPLITS['closes_test'], SPLITS['retornos_test'],
    SPLITS['datas_test'],
    CAPITAL_INICIAL, PROB_THRESHOLD,
    RISCO_POR_TRADE, STOP_LOSS_PCT, TAKE_PROFIT_PCT,
    SPREAD_PIPS, SLIPPAGE
)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║             PASSO 15 — FEATURE IMPORTANCE + SHAP                        ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def plotar_feature_importance(modelo, feature_names: list,
                               titulo: str = "Feature Importance") -> None:
    """
    Plota a importância das features do modelo (se disponível) e SHAP values.
    """
    print("\n" + "─" * 60)
    print("  PASSO 15 — Feature Importance + SHAP")
    print("─" * 60)

    has_fi = hasattr(modelo, 'feature_importances_')

    if not has_fi and not SHAP_AVAILABLE:
        print("  Modelo não suporta feature importance direta nem SHAP.")
        return

    if USAR_PCA:
        print("  [INFO] PCA ativo — interpretabilidade por features originais não disponível.")
        return

    fig, axes = plt.subplots(1, 2 if SHAP_AVAILABLE else 1,
                              figsize=(16 if SHAP_AVAILABLE else 10, 8))
    if not SHAP_AVAILABLE:
        axes = [axes]

    # ── Feature Importance do Modelo ──────────────────────────────────────────
    if has_fi:
        ax = axes[0]
        importances = pd.Series(modelo.feature_importances_,
                                index=feature_names).sort_values(ascending=True)
        top_n = importances.tail(25)
        colors_fi = [COLORS['primary'] if i >= len(importances) - 5
                     else COLORS['neutral'] for i in range(len(top_n))]
        top_n.plot(kind='barh', ax=ax, color=colors_fi)
        ax.set_title(f'{titulo}\n(Top 25)', fontweight='bold')
        ax.set_xlabel('Importância')
        ax.axvline(x=0, color=COLORS['border'], linestyle='-', linewidth=0.5)

    # ── SHAP Values ───────────────────────────────────────────────────────────
    if SHAP_AVAILABLE and not USAR_PCA:
        print("  Calculando SHAP values...")
        ax_shap = axes[1] if len(axes) > 1 else axes[0]

        try:
            X_shap = X_test_s[:200]  # Amostra para velocidade

            # ── Escolher o explainer correto pelo tipo do modelo ──────────
            nome_classe = type(modelo).__name__
            tree_classes = ('XGBClassifier', 'LGBMClassifier',
                            'CatBoostClassifier', 'RandomForestClassifier',
                            'DecisionTreeClassifier', 'GradientBoostingClassifier')
            linear_classes = ('LogisticRegression', 'LinearSVC',
                              'SGDClassifier', 'RidgeClassifier')

            if nome_classe in tree_classes:
                explainer = shap.TreeExplainer(modelo)
                shap_vals = explainer.shap_values(X_shap)
            elif nome_classe in linear_classes:
                # LinearExplainer precisa de background data (mediana do treino)
                bg = shap.maskers.Independent(X_train_s, max_samples=100)
                explainer = shap.LinearExplainer(modelo, bg)
                shap_vals = explainer.shap_values(X_shap)
            else:
                # KernelExplainer: universal mas lento — usa amostra pequena
                print("  [SHAP] Usando KernelExplainer (pode demorar ~1min)...")
                bg_k = shap.sample(X_train_s, 50)
                explainer = shap.KernelExplainer(modelo.predict_proba, bg_k)
                shap_vals = explainer.shap_values(X_shap[:50])

            # Para classificação binária, pegar classe 1
            if isinstance(shap_vals, list) and len(shap_vals) == 2:
                shap_vals = shap_vals[1]
            elif hasattr(shap_vals, 'shape') and len(shap_vals.shape) == 3:
                shap_vals = shap_vals[:, :, 1]

            # SHAP summary plot
            plt.figure(figsize=(10, 8))
            shap.summary_plot(
                shap_vals, X_shap,
                feature_names=feature_names,
                plot_type='bar',
                show=False,
                max_display=20,
            )
            plt.title(f'SHAP Feature Importance — {nome_classe}',
                      fontweight='bold', color=COLORS['text_main'])
            plt.tight_layout()
            plt.savefig(f'{_COLAB_OUTPUT_DIR}/shap_importance.png',
                        dpi=120, bbox_inches='tight', facecolor=COLORS['bg_dark'])
            plt.show()
            print(f"  SHAP calculado com sucesso ({nome_classe}).")
        except Exception as e:
            print(f"  [AVISO] SHAP falhou: {e}")

    plt.tight_layout()
    plt.savefig(f'{_COLAB_OUTPUT_DIR}/feature_importance.png',
                dpi=120, bbox_inches='tight', facecolor=COLORS['bg_dark'])
    plt.show()

plotar_feature_importance(MELHOR_MODELO, FEATURES_SELECIONADAS, MELHOR_MODELO_NOME)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║          PASSO 16 — PREVISÃO COM THRESHOLD DE PROBABILIDADE             ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def analisar_threshold(modelo, X_test, y_test, retornos_test) -> dict:
    """
    Analisa como diferentes thresholds de probabilidade afetam as métricas.
    Retorna o threshold ótimo baseado no F1-Score.
    """
    print("\n" + "─" * 60)
    print("  PASSO 16 — Análise de Threshold de Probabilidade")
    print("─" * 60)

    y_prob = modelo.predict_proba(X_test)[:, 1]
    thresholds = np.arange(0.40, 0.75, 0.02)

    resultados = []
    for thr in thresholds:
        y_pred_thr = (y_prob >= thr).astype(int)
        n_sinais = y_pred_thr.sum()
        if n_sinais < 5:
            continue

        f1   = f1_score(y_test, y_pred_thr, zero_division=0)
        acc  = accuracy_score(y_test, y_pred_thr)
        prec = precision_score(y_test, y_pred_thr, zero_division=0)
        rec  = recall_score(y_test, y_pred_thr, zero_division=0)

        # Retorno da estratégia
        sinais = np.where(y_pred_thr == 1, 1, 0)
        rets_est = sinais * retornos_test / 100
        ret_total = (1 + rets_est).prod() - 1 if len(rets_est) > 0 else 0

        resultados.append({
            'threshold': thr,
            'f1': f1, 'acc': acc, 'prec': prec, 'rec': rec,
            'n_sinais': n_sinais,
            'pct_sinais': n_sinais / len(y_test),
            'retorno': ret_total,
        })

    df_thr = pd.DataFrame(resultados)
    if len(df_thr) == 0:
        return {'threshold_otimo': PROB_THRESHOLD}

    # Threshold ótimo = maior F1
    idx_otimo = df_thr['f1'].idxmax()
    thr_otimo = df_thr.loc[idx_otimo, 'threshold']

    print(f"  Threshold ótimo (F1): {thr_otimo:.2f}")
    print(f"  F1 no threshold ótimo: {df_thr.loc[idx_otimo, 'f1']:.4f}")
    print(f"  Sinais gerados: {df_thr.loc[idx_otimo, 'n_sinais']:,} "
          f"({df_thr.loc[idx_otimo, 'pct_sinais']:.1%} do período)")

    # Gráfico threshold vs métricas
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f'Análise de Threshold | {TICKER}', fontsize=13,
                 fontweight='bold', color=COLORS['text_main'])

    ax1.plot(df_thr['threshold'], df_thr['f1'],    color=COLORS['primary'],   label='F1', linewidth=2)
    ax1.plot(df_thr['threshold'], df_thr['prec'],  color=COLORS['success'],   label='Precision', linewidth=2)
    ax1.plot(df_thr['threshold'], df_thr['rec'],   color=COLORS['warning'],   label='Recall', linewidth=2)
    ax1.plot(df_thr['threshold'], df_thr['acc'],   color=COLORS['highlight'], label='Accuracy', linewidth=2)
    ax1.axvline(thr_otimo, color=COLORS['danger'], linestyle='--', label=f'Ótimo={thr_otimo:.2f}')
    ax1.set_title('Métricas vs Threshold', color=COLORS['text_main'])
    ax1.set_xlabel('Threshold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    ax2.bar(df_thr['threshold'], df_thr['pct_sinais'] * 100,
            color=COLORS['primary'], alpha=0.7, width=0.015)
    ax2.set_title('% de Sinais Gerados por Threshold', color=COLORS['text_main'])
    ax2.set_xlabel('Threshold')
    ax2.set_ylabel('% Sinais')

    plt.tight_layout()
    plt.savefig(f'{_COLAB_OUTPUT_DIR}/threshold_analysis.png',
                dpi=120, bbox_inches='tight', facecolor=COLORS['bg_dark'])
    plt.show()

    return {'threshold_otimo': thr_otimo, 'df_threshold': df_thr}

THRESHOLD_ANALISE = analisar_threshold(
    MELHOR_MODELO, X_test_s,
    SPLITS['y_test'], SPLITS['retornos_test']
)

# ── Atualizar PROB_THRESHOLD com o valor ótimo calculado ─────────────
# O Passo 16 encontrou o threshold que maximiza o F1-Score.
# Usamos esse valor daqui em diante (backtest, walk-forward, produção).
if 'threshold_otimo' in THRESHOLD_ANALISE:
    PROB_THRESHOLD = float(THRESHOLD_ANALISE['threshold_otimo'])
    print(f"  [AUTO] PROB_THRESHOLD atualizado para {PROB_THRESHOLD:.2f} "
          f"(threshold ótimo do Passo 16)")


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                 PASSO 17 — ANÁLISE DE OVERFITTING                       ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Overfitting:
#   Ocorre quando o modelo "memoriza" os dados de treino em vez de aprender
#   padrões generalizáveis. Sinais de overfitting:
#   1. Acc_Train >> Acc_Test (gap grande)
#   2. ROC-AUC cai muito de treino para teste
#   3. Curva de aprendizado não converge
#   Soluções: regularização, early stopping, reduzir complexidade,
#             mais dados, dropout, cross-validation rigorosa.

def analisar_overfitting(modelo, nome: str,
                          X_train, y_train,
                          X_val, y_val,
                          X_test, y_test) -> None:
    """
    Compara performance em treino, validação e teste para detectar overfitting.
    """
    print("\n" + "─" * 60)
    print("  PASSO 17 — Análise de Overfitting")
    print("─" * 60)

    conjuntos = {
        'Treino':    (X_train, y_train),
        'Validação': (X_val,   y_val),
        'Teste':     (X_test,  y_test),
    }

    metricas_ov = {}
    for nome_conj, (X, y) in conjuntos.items():
        y_pred = modelo.predict(X)
        y_prob = modelo.predict_proba(X)[:, 1]
        metricas_ov[nome_conj] = {
            'Accuracy':  accuracy_score(y, y_pred),
            'F1':        f1_score(y, y_pred, zero_division=0),
            'ROC-AUC':   roc_auc_score(y, y_prob),
        }

    df_ov = pd.DataFrame(metricas_ov).T
    print(df_ov.to_string(float_format='{:.4f}'.format))

    # Diagnóstico
    gap_roc = metricas_ov['Treino']['ROC-AUC'] - metricas_ov['Teste']['ROC-AUC']
    gap_acc = metricas_ov['Treino']['Accuracy'] - metricas_ov['Teste']['Accuracy']

    print(f"\n  Gap ROC-AUC (Treino - Teste) : {gap_roc:.4f}")
    print(f"  Gap Accuracy (Treino - Teste) : {gap_acc:.4f}")

    if gap_roc > 0.10:
        print("  [⚠ ALERTA] Gap > 0.10 — Possível overfitting significativo!")
        print("  → Reduza max_depth, aumente min_samples_leaf, adicione regularização.")
    elif gap_roc > 0.05:
        print("  [~ ATENÇÃO] Gap moderado (0.05–0.10) — Monitorar.")
    else:
        print("  [✓ OK] Gap pequeno — Modelo generalizando bem.")

    # Gráfico
    fig, ax = plt.subplots(figsize=(8, 5))
    fig.suptitle(f'Overfitting Analysis | {nome} | {TICKER}',
                 fontsize=13, fontweight='bold', color=COLORS['text_main'])

    metricas_plot = ['Accuracy', 'F1', 'ROC-AUC']
    x = np.arange(len(metricas_plot))
    width = 0.25
    palette = [COLORS['primary'], COLORS['warning'], COLORS['success']]

    for i, (conj_nome, vals) in enumerate(metricas_ov.items()):
        vals_list = [vals[m] for m in metricas_plot]
        bars = ax.bar(x + i * width, vals_list, width, label=conj_nome,
                      color=palette[i], alpha=0.85)
        for bar, v in zip(bars, vals_list):
            ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.005,
                    f'{v:.3f}', ha='center', va='bottom', fontsize=8,
                    color=COLORS['text_main'])

    ax.set_xticks(x + width)
    ax.set_xticklabels(metricas_plot)
    ax.set_ylim(0, 1.1)
    ax.legend()
    ax.set_title('Treino vs Validação vs Teste', color=COLORS['text_main'])
    ax.set_ylabel('Score')
    ax.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'{_COLAB_OUTPUT_DIR}/overfitting_analysis.png',
                dpi=120, bbox_inches='tight', facecolor=COLORS['bg_dark'])
    plt.show()

analisar_overfitting(
    MELHOR_MODELO, MELHOR_MODELO_NOME,
    X_train_s, SPLITS['y_train'],
    X_val_s,   SPLITS['y_val'],
    X_test_s,  SPLITS['y_test']
)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║              PASSO 18 — WALK-FORWARD OPTIMIZATION                       ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================
# TEORIA — Walk-Forward Validation:
#   Simula o processo real de trading: treinamos com dados históricos,
#   fazemos previsões nos dados imediatamente seguintes (OOS), depois
#   avançamos a janela e repetimos. Isso elimina viés de seleção e
#   mede performance out-of-sample de forma realista.
#
#   Expanding Window:
#   ├── Window 1: Treino [0:200]   → Teste [200:260]
#   ├── Window 2: Treino [0:260]   → Teste [260:320]  (treino CRESCE)
#   └── Window N: Treino [0:N*60]  → Teste [N*60:(N+1)*60]

def walk_forward_validation(X: np.ndarray, y: np.ndarray,
                              datas, retornos: np.ndarray,
                              n_splits: int = 5,
                              test_size: int = 60) -> dict:
    """
    Walk-forward validation com expanding window.
    Retreina o modelo em cada janela para simular uso real.
    """
    print("\n" + "─" * 60)
    print("  PASSO 18 — Walk-Forward Validation")
    print("─" * 60)

    tscv = TimeSeriesSplit(n_splits=n_splits, test_size=test_size)

    wf_resultados = []
    retornos_wf_todos = []

    for fold, (train_idx, test_idx) in enumerate(tscv.split(X), 1):
        X_tr, X_te = X[train_idx], X[test_idx]
        y_tr, y_te = y[train_idx], y[test_idx]
        ret_te = retornos[test_idx]

        # Normalizar APENAS com dados do treino desta janela
        sc = RobustScaler()
        X_tr_s = sc.fit_transform(X_tr)
        X_te_s = sc.transform(X_te)

        # Treinar modelo fresh (sem memorizar janelas anteriores)
        if XGB_AVAILABLE:
            modelo_wf = XGBClassifier(
                n_estimators=200, max_depth=4, learning_rate=0.05,
                subsample=0.8,
                eval_metric='logloss', random_state=42,
                n_jobs=-1, verbosity=0
            )
        else:
            modelo_wf = RandomForestClassifier(
                n_estimators=100, max_depth=5, random_state=42, n_jobs=-1
            )

        modelo_wf.fit(X_tr_s, y_tr)
        y_pred = modelo_wf.predict(X_te_s)
        y_prob = modelo_wf.predict_proba(X_te_s)[:, 1]

        # Métricas desta janela
        acc = accuracy_score(y_te, y_pred)
        roc = roc_auc_score(y_te, y_prob)
        f1  = f1_score(y_te, y_pred, zero_division=0)

        # Retorno da estratégia nesta janela
        sinais = np.where(y_prob >= PROB_THRESHOLD, 1,
                          np.where(y_prob <= 1 - PROB_THRESHOLD, -1, 0))
        ret_est = sinais * ret_te / 100
        ret_total = (1 + ret_est).prod() - 1

        dt_inicio = datas[test_idx[0]].date()
        dt_fim    = datas[test_idx[-1]].date()

        wf_resultados.append({
            'Fold': fold,
            'Período': f"{dt_inicio} → {dt_fim}",
            'N_Treino': len(train_idx),
            'N_Teste': len(test_idx),
            'Accuracy': acc,
            'ROC-AUC': roc,
            'F1': f1,
            'Retorno': ret_total,
        })
        retornos_wf_todos.extend(ret_est.tolist())

        print(f"  Fold {fold}: {dt_inicio} → {dt_fim} | "
              f"ROC={roc:.4f} | Acc={acc:.4f} | Ret={ret_total:.2%}")

    df_wf = pd.DataFrame(wf_resultados)
    retornos_wf_series = pd.Series(retornos_wf_todos)

    print(f"\n  ── Walk-Forward Médias ──")
    print(f"  ROC-AUC médio  : {df_wf['ROC-AUC'].mean():.4f} ± {df_wf['ROC-AUC'].std():.4f}")
    print(f"  Accuracy médio : {df_wf['Accuracy'].mean():.4f} ± {df_wf['Accuracy'].std():.4f}")
    print(f"  Retorno médio  : {df_wf['Retorno'].mean():.2%} ± {df_wf['Retorno'].std():.2%}")
    print(f"  Retorno total  : {(1 + retornos_wf_series).prod() - 1:.2%}")

    return {'df_wf': df_wf, 'retornos_todos': retornos_wf_series}

WF_RESULTADOS = walk_forward_validation(
    X_all_s, y, DATAS, RETORNOS,
    WF_N_SPLITS, WF_TEST_SIZE
)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                  PASSO 19 — GRÁFICOS PROFISSIONAIS                      ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def gerar_graficos_finais(avaliacao: dict, backtest: dict,
                           df_resultados: pd.DataFrame,
                           splits: dict, ticker: str) -> None:
    """
    Gera o painel visual completo de resultados.
    """
    print("\n" + "─" * 60)
    print("  PASSO 19 — Gerando Gráficos Profissionais")
    print("─" * 60)

    metricas_fin = avaliacao['metricas_fin']
    equity_mod = metricas_fin['equity']
    equity_bh  = metricas_fin['equity_bh']
    drawdown   = metricas_fin['drawdown']
    datas_test = splits['datas_test']

    # ── Figura 1: Equity Curve + Drawdown ─────────────────────────────────────
    fig = plt.figure(figsize=(16, 12))
    fig.suptitle(f'PAINEL DE RESULTADOS — {ticker} | {MELHOR_MODELO_NOME}',
                 fontsize=15, fontweight='bold', color=COLORS['text_main'])

    gs = gridspec.GridSpec(3, 3, figure=fig, hspace=0.45, wspace=0.35)

    # Equity Curve
    ax1 = fig.add_subplot(gs[0, :2])
    ax1.plot(datas_test[:len(equity_mod)], equity_mod.values,
             color=COLORS['primary'], linewidth=2, label='Modelo', zorder=3)
    ax1.plot(datas_test[:len(equity_bh)], equity_bh.values,
             color=COLORS['warning'], linewidth=1.5, linestyle='--',
             label='Buy & Hold', alpha=0.8)
    ax1.fill_between(datas_test[:len(equity_mod)], equity_mod.values,
                     CAPITAL_INICIAL, alpha=0.08, color=COLORS['primary'])
    ax1.axhline(CAPITAL_INICIAL, color=COLORS['border'], linestyle=':', alpha=0.5)
    ax1.set_title('Equity Curve', color=COLORS['text_main'], fontweight='bold')
    ax1.set_ylabel('Capital (USD)')
    ax1.legend()
    ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:,.0f}'))
    ax1.grid(True, alpha=0.2)

    # Drawdown
    ax2 = fig.add_subplot(gs[1, :2])
    dd_values = drawdown.values
    ax2.fill_between(range(len(dd_values)), dd_values * 100, 0,
                     color=COLORS['danger'], alpha=0.6)
    ax2.plot(range(len(dd_values)), dd_values * 100,
             color=COLORS['danger'], linewidth=1)
    ax2.set_title('Drawdown (%)', color=COLORS['text_main'], fontweight='bold')
    ax2.set_ylabel('%')
    ax2.grid(True, alpha=0.2)

    # Comparação de modelos (ROC-AUC)
    ax3 = fig.add_subplot(gs[0, 2])
    df_plot = df_resultados.sort_values('ROC_Test')
    colors_bar = [COLORS['primary'] if n == MELHOR_MODELO_NOME
                  else COLORS['neutral'] for n in df_plot['Modelo']]
    ax3.barh(df_plot['Modelo'], df_plot['ROC_Test'], color=colors_bar, alpha=0.85)
    ax3.axvline(0.5, color=COLORS['danger'], linestyle='--', alpha=0.7, label='Random')
    ax3.set_title('ROC-AUC por Modelo', color=COLORS['text_main'], fontweight='bold')
    ax3.set_xlabel('ROC-AUC')
    for i, v in enumerate(df_plot['ROC_Test']):
        ax3.text(v + 0.001, i, f'{v:.3f}', va='center', fontsize=8,
                 color=COLORS['text_main'])
    ax3.grid(axis='x', alpha=0.3)
    ax3.legend(fontsize=8)

    # Distribuição dos retornos
    ax4 = fig.add_subplot(gs[1, 2])
    rets = SPLITS['retornos_test']
    ax4.hist(rets, bins=50, color=COLORS['primary'], alpha=0.7,
             edgecolor='none', density=True)
    mu, std = rets.mean(), rets.std()
    x_norm = np.linspace(mu - 4*std, mu + 4*std, 100)
    from scipy.stats import norm as sci_norm
    ax4.plot(x_norm, sci_norm.pdf(x_norm, mu, std),
             color=COLORS['warning'], linewidth=2, label='Normal')
    ax4.axvline(0, color=COLORS['danger'], linestyle='--', alpha=0.7)
    ax4.set_title('Distribuição Retornos', color=COLORS['text_main'], fontweight='bold')
    ax4.set_xlabel('%')
    ax4.legend(fontsize=8)

    # ROC Curve
    ax5 = fig.add_subplot(gs[2, :2])
    fpr, tpr, _ = roc_curve(SPLITS['y_test'], AVALIACAO['y_prob'])
    auc = AVALIACAO['roc']
    ax5.plot(fpr, tpr, color=COLORS['primary'], linewidth=2,
             label=f'ROC-AUC = {auc:.4f}')
    ax5.plot([0, 1], [0, 1], color=COLORS['neutral'], linestyle='--',
             alpha=0.5, label='Random (0.5)')
    ax5.fill_between(fpr, tpr, alpha=0.1, color=COLORS['primary'])
    ax5.set_title('ROC Curve', color=COLORS['text_main'], fontweight='bold')
    ax5.set_xlabel('False Positive Rate')
    ax5.set_ylabel('True Positive Rate')
    ax5.legend()
    ax5.grid(True, alpha=0.2)

    # Confusion Matrix
    ax6 = fig.add_subplot(gs[2, 2])
    cm = confusion_matrix(SPLITS['y_test'], AVALIACAO['y_pred'])
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax6,
                xticklabels=['Queda', 'Alta'],
                yticklabels=['Queda', 'Alta'],
                linewidths=1, linecolor=COLORS['border'])
    ax6.set_title('Confusion Matrix', color=COLORS['text_main'], fontweight='bold')
    ax6.set_xlabel('Predito')
    ax6.set_ylabel('Real')

    plt.savefig(f'{_COLAB_OUTPUT_DIR}/painel_resultados.png',
                dpi=140, bbox_inches='tight', facecolor=COLORS['bg_dark'])
    plt.show()
    print("  Painel salvo em outputs/painel_resultados.png")

    # ── Figura 2: Heatmap de Correlação ───────────────────────────────────────
    print("  Gerando heatmap de correlação...")
    top_features = list(MI_RANKING.head(20).index)
    df_corr = df[top_features].dropna().corr()

    fig2, ax_corr = plt.subplots(figsize=(13, 10))
    fig2.suptitle(f'Correlação — Top 20 Features | {ticker}',
                  fontsize=13, fontweight='bold', color=COLORS['text_main'])
    mask = np.triu(np.ones_like(df_corr, dtype=bool))
    sns.heatmap(df_corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
                vmin=-1, vmax=1, center=0, ax=ax_corr,
                linewidths=0.5, linecolor=COLORS['border'],
                annot_kws={'size': 7}, square=True)
    ax_corr.tick_params(axis='x', rotation=45, labelsize=8)
    ax_corr.tick_params(axis='y', labelsize=8)
    plt.tight_layout()
    plt.savefig(f'{_COLAB_OUTPUT_DIR}/correlacao_features.png',
                dpi=120, bbox_inches='tight', facecolor=COLORS['bg_dark'])
    plt.show()
    print("  Heatmap salvo em outputs/correlacao_features.png")

gerar_graficos_finais(AVALIACAO, BACKTEST, DF_RESULTADOS, SPLITS, TICKER)


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║          PASSO 20 — PREVISÃO PARA O PRÓXIMO DIA (PRODUÇÃO)              ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def predict_next_day(ticker: str = TICKER,
                      modelo=MELHOR_MODELO,
                      scaler_obj=scaler,
                      feature_names: list = FEATURES_SELECIONADAS,
                      threshold: float = PROB_THRESHOLD) -> dict:
    """
    Função de produção: baixa o dado mais recente, recalcula indicadores
    e gera a previsão para o próximo dia de negociação.

    Retorno: dict com sinal, probabilidade, confiança e recomendação
    """
    print("\n" + "=" * 60)
    print(f"  PASSO 20 — PREVISÃO PARA O PRÓXIMO DIA | {ticker}")
    print("=" * 60)

    # Download dos dados mais recentes (90 dias para calcular indicadores)
    df_novo = baixar_dados(ticker, anos=1, interval="1d")
    df_novo = limpar_dados(df_novo)
    df_novo = construir_target(df_novo)
    df_novo = engenharia_de_features(df_novo)

    # Verificar se todas as features existem
    features_ok = [f for f in feature_names if f in df_novo.columns]
    features_faltando = [f for f in feature_names if f not in df_novo.columns]

    if len(features_faltando) > 0:
        print(f"  [AVISO] {len(features_faltando)} features ausentes no dado recente.")
        # Preencher faltantes com 0
        for f in features_faltando:
            df_novo[f] = 0.0

    # Última linha = dado mais recente (para prever amanhã)
    ultima_linha = df_novo[feature_names].iloc[-1:]

    if ultima_linha.isnull().any().any():
        ultima_linha = ultima_linha.fillna(0)

    # Normalizar com o scaler treinado
    X_prod = scaler_obj.transform(ultima_linha.values)

    if USAR_PCA and 'pca_final' in dir():
        X_prod = pca_final.transform(X_prod)

    # Predição
    prob_alta = modelo.predict_proba(X_prod)[0, 1]
    prob_queda = 1 - prob_alta

    # Definir sinal
    if prob_alta >= threshold:
        sinal = "COMPRA 🟢"
        sinal_code = "BUY"
        confianca = prob_alta
    elif prob_queda >= threshold:
        sinal = "VENDA 🔴"
        sinal_code = "SELL"
        confianca = prob_queda
    else:
        sinal = "NEUTRO ⚪"
        sinal_code = "NEUTRAL"
        confianca = max(prob_alta, prob_queda)

    # Dados de referência
    preco_atual    = df_novo['Close'].iloc[-1]
    retorno_hoje   = df_novo['retorno_atual'].iloc[-1] * 100
    vol_20     = df_novo['vol_20'].iloc[-1] if 'vol_20' in df_novo.columns else 0.0
    rsi_atual  = df_novo['rsi_14'].iloc[-1] if 'rsi_14' in df_novo.columns else 50.0
    data_referencia = df_novo.index[-1].date()

    # Exibir resultado
    print(f"\n  {'─'*50}")
    print(f"  Data de referência : {data_referencia}")
    print(f"  Preço atual ({ticker}) : {preco_atual:.4f}")
    print(f"  Retorno hoje       : {retorno_hoje:+.3f}%")
    print(f"  Volatilidade (20d) : {vol_20:.4f}")
    print(f"  RSI (14)           : {rsi_atual:.2f}")
    print(f"  {'─'*50}")
    print(f"  Prob. Alta         : {prob_alta:.4f}  ({prob_alta:.1%})")
    print(f"  Prob. Queda        : {prob_queda:.4f}  ({prob_queda:.1%})")
    print(f"  Threshold          : {threshold:.0%}")
    print(f"  {'─'*50}")
    print(f"\n  ▶ SINAL: {sinal}")
    print(f"  ▶ CONFIANÇA: {confianca:.2%}")
    print(f"  {'─'*50}")

    if sinal_code == "BUY":
        print(f"\n  💡 Interpretação: Modelo espera ALTA para amanhã com "
              f"{prob_alta:.1%} de confiança.")
        print(f"  ⚠ Stop-Loss sugerido: -{STOP_LOSS_PCT:.1%} | "
              f"Take-Profit: +{TAKE_PROFIT_PCT:.1%}")
    elif sinal_code == "SELL":
        print(f"\n  💡 Interpretação: Modelo espera QUEDA para amanhã com "
              f"{prob_queda:.1%} de confiança.")
        print(f"  ⚠ Stop-Loss sugerido: -{STOP_LOSS_PCT:.1%} | "
              f"Take-Profit: +{TAKE_PROFIT_PCT:.1%}")
    else:
        print(f"\n  💡 Interpretação: Probabilidade insuficiente ({confianca:.1%}) "
              f"para operar. Aguardar melhor oportunidade.")

    print(f"\n  ⚠ AVISO LEGAL: Esta previsão é para fins educacionais/pesquisa.")
    print(f"  NÃO constitui recomendação financeira. Consulte um profissional.")

    return {
        'ticker': ticker,
        'data_referencia': str(data_referencia),
        'preco_atual': preco_atual,
        'prob_alta': prob_alta,
        'prob_queda': prob_queda,
        'sinal': sinal_code,
        'confianca': confianca,
        'threshold': threshold,
    }

# ─── Executar previsão ────────────────────────────────────────────────────────
PREVISAO = predict_next_day()


# ==============================================================================
# ╔══════════════════════════════════════════════════════════════════════════╗
# ║                  RESUMO FINAL E RANKING DOS MODELOS                     ║
# ╚══════════════════════════════════════════════════════════════════════════╝
# ==============================================================================

def exibir_resumo_final(df_resultados: pd.DataFrame,
                         avaliacao: dict,
                         backtest: dict,
                         wf: dict,
                         previsao: dict) -> None:
    """
    Exibe o resumo final completo do sistema.
    """
    print("\n")
    print("=" * 70)
    print("  RESUMO FINAL — SISTEMA QUANTITATIVO DE ML PARA TRADING")
    print("=" * 70)

    print(f"\n  Ativo           : {TICKER}")
    print(f"  Período dados   : {DATAS[0].date()} → {DATAS[-1].date()}")
    print(f"  Total amostras  : {len(y):,}")
    print(f"  Features usadas : {len(FEATURES_SELECIONADAS)}")

    print(f"\n  {'─'*60}")
    print(f"  RANKING DOS MODELOS (por ROC-AUC no Teste)")
    print(f"  {'─'*60}")
    print(f"  {'#':<4} {'Modelo':<16} {'ROC-AUC':<10} {'Acc':<10} {'F1':<10}")
    print(f"  {'─'*60}")
    for i, (_, row) in enumerate(df_resultados.iterrows(), 1):
        marker = " ◀ MELHOR" if i == 1 else ""
        print(f"  {i:<4} {row['Modelo']:<16} {row['ROC_Test']:<10.4f} "
              f"{row['Acc_Test']:<10.4f} {row['F1_Test']:<10.4f}{marker}")

    print(f"\n  {'─'*60}")
    print(f"  PERFORMANCE FINANCEIRA — Modelo: {MELHOR_MODELO_NOME}")
    print(f"  {'─'*60}")
    mf = avaliacao['metricas_fin']
    print(f"  Retorno Modelo  : {mf['Retorno Total']:>10.2%}")
    print(f"  Retorno B&H     : {mf['Retorno B&H']:>10.2%}")
    print(f"  Sharpe Ratio    : {mf['Sharpe Ratio']:>10.3f}")
    print(f"  Sortino Ratio   : {mf['Sortino Ratio']:>10.3f}")
    print(f"  Max Drawdown    : {mf['Max Drawdown']:>10.2%}")
    print(f"  Win Rate        : {mf['Win Rate']:>10.2%}")
    print(f"  Profit Factor   : {mf['Profit Factor']:>10.3f}")
    print(f"  Calmar Ratio    : {mf['Calmar Ratio']:>10.3f}")
    print(f"  Expectancy      : {mf['Expectancy']:>10.4f}")

    print(f"\n  {'─'*60}")
    print(f"  BACKTEST REALISTA (Threshold: {PROB_THRESHOLD:.0%})")
    print(f"  {'─'*60}")
    print(f"  Capital Final   : ${backtest['capital_final']:>12,.2f}")
    print(f"  Retorno Total   : {backtest['ret_total']:>10.2%}")
    print(f"  Nº Trades       : {backtest['n_trades']:>10,}")
    print(f"  Win Rate        : {backtest['win_rate']:>10.2%}")
    print(f"  Profit Factor   : {backtest['profit_factor']:>10.3f}")
    print(f"  Max Drawdown    : {backtest['max_dd']:>10.2%}")

    print(f"\n  {'─'*60}")
    print(f"  WALK-FORWARD (OOS médio)")
    print(f"  {'─'*60}")
    df_wf = wf['df_wf']
    print(f"  ROC-AUC médio   : {df_wf['ROC-AUC'].mean():.4f} ± {df_wf['ROC-AUC'].std():.4f}")
    print(f"  Retorno médio   : {df_wf['Retorno'].mean():.2%} ± {df_wf['Retorno'].std():.2%}")

    print(f"\n  {'─'*60}")
    print(f"  SINAL ATUAL PARA PRÓXIMO DIA")
    print(f"  {'─'*60}")
    emojis = {'BUY': '🟢 COMPRA', 'SELL': '🔴 VENDA', 'NEUTRAL': '⚪ NEUTRO'}
    print(f"  Sinal           : {emojis.get(previsao['sinal'], previsao['sinal'])}")
    print(f"  Probabilidade ↑ : {previsao['prob_alta']:.2%}")
    print(f"  Probabilidade ↓ : {previsao['prob_queda']:.2%}")
    print(f"  Confiança       : {previsao['confianca']:.2%}")
    print(f"  Data referência : {previsao['data_referencia']}")

    print(f"\n  {'─'*60}")
    print(f"  ARQUIVOS GERADOS (em outputs/)")
    print(f"  {'─'*60}")
    arquivos = [
        'painel_resultados.png',
        'correlacao_features.png',
        'feature_importance.png',
        'threshold_analysis.png',
        'overfitting_analysis.png',
    ]
    if SHAP_AVAILABLE and not USAR_PCA:
        arquivos.append('shap_importance.png')
    if USAR_PCA:
        arquivos.append('pca_scree.png')

    for arq in arquivos:
        path = f'{_COLAB_OUTPUT_DIR}/{arq}'
        existe = os.path.exists(path)
        print(f"  {'✓' if existe else '○'} {arq}")

    print("\n" + "=" * 70)
    print("  FIM DO SISTEMA QUANTITATIVO DE ML PARA TRADING")
    print("=" * 70)
    print("\n  ⚠ DISCLAIMER: Resultados passados não garantem resultados futuros.")
    print("  Este sistema é para fins educacionais e de pesquisa quantitativa.")
    print("  NÃO constitui recomendação de investimento.")

exibir_resumo_final(DF_RESULTADOS, AVALIACAO, BACKTEST, WF_RESULTADOS, PREVISAO)
```
